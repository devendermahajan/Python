# **INNER FUNCTION**
---
An inner function is simply a function defined inside another function. Inner functions are not accessible outside their containing (outer) function and have their own local scope. They are commonly used for organizational purposes, encapsulation, or creating closures. It also known as *nested function*.

### **ADVANTAGES OF USER-DEFINED FUNCTIONS**
1. **Encapsulation:** Inner functions allow to encapsulate logic that is only relevant within the outer function. This helps in organizing code and avoiding unnecessary exposure of functionality.

2. **Code Reusability:** It helps defining reusable helper functions within a larger function to simplify complex operations.

3. **Closures:** Inner functions can "remember" the variables from their enclosing scope, enabling powerful programming techniques like decorators.

4. **Improved Readability:** By grouping related logic inside an outer function, inner functions can make the code more readable and maintainable.

## **FUNCTION DEFINITION**

A function can be create using following syntax: 

```python
def outer-function-name ([parameters]): 
    # function body
    def inner-function-name ([parameters]):
        # logic for inner function

    # logic for outer function
```

````python
# simple nested function
def greet_user(user):
    def greet():
        return f"Hello, {user}!"
    return greet()
````

````python
print(greet_user('Python'))
````

### **Encapulation**
Encapsulation refers to keeping related functionality together within an outer function. Inner functions ensure that their logic is only accessible where it is relevant, reducing potential conflicts with global or module-level code.

````python
def calculate_area(shape, dimension):
    def circle_area(radius):
        return 3.14 * radius * radius

    def square_area(side):
        return side * side

    if shape == "circle":
        return circle_area(dimension)
    elif shape == "square":
        return square_area(dimension)
    else:
        return "Unknown shape"

print(calculate_area("circle", 5))  # Output: 78.5
print(calculate_area("square", 4))  # Output: 16
````

### **Using Closures**

Closures are a way for inner functions to “remember” the variables from their enclosing scope, even after the outer function has finished executing. Closures can be particularly useful in scenarios like creating customized functions, maintaining state between function calls, or building decorators.

### **Conditions to create a closure**:

- There must be a nested function.

- The inner function has to refer to a value that is defined in the enclosing scope.

- The enclosing function has to return the nested function.

Closures helps in defining specific behaviors like squaring or cubing in a highly modular and reusable manner, as shown below

````python
def power(exponent):
    def raise_to_power(base):
        return base ** exponent

    return raise_to_power

square = power(2)
cube = power(3)

print(square(4))  # Output: 16
print(cube(2))    # Output: 8
````

### **Decorators**

Inner functions are heavily used in Python decorators, which modify the behavior of a function. Decorators are a powerful feature of Python that leverage inner functions to wrap additional functionality around an existing function.

In below example, the `decorator` function takes another function as an argument (`func`). It defines a nested wrapper function that adds behavior before and after the execution of `func`. The `@decorator` syntax is a shorthand for wrapping the `say_hello` function with `decorator`.

````python
def decorator(func):
    def wrapper():
        print("Something is happening before the function is called.")
        func()
        print("Something is happening after the function is called.")

    return wrapper

@decorator
def say_hello():
    print("Hello!")

say_hello()
````

> In the example below, the `repeat` function is a decorator factory that accepts an argument (`n`) and returns a decorator. *Hence, inner functions make decorators highly customizable and dynamic*.

````python
def repeat(n):
    def decorator(func):
        def wrapper():
            for _ in range(n):
                func()
        return wrapper
    return decorator

@repeat(3)
def greet():
    print("Hello!")

greet()
````

### **Dynamic Function Creation**

Inner functions can dynamically generate and return new functions based on logic within the outer function. This is useful for scenarios requiring programmatically defined behaviors.

````python
def create_math_operation(operation):
    def add(a, b):
        return a + b

    def subtract(a, b):
        return a - b

    if operation == "add":
        return add
    elif operation == "subtract":
        return subtract
    else:
        raise ValueError("Unsupported operation")

sigma = create_math_operation("add")
print(sigma(10, 5))  # Output: 15

difference = create_math_operation("subtract")
print(difference(10, 5))  # Output: 5
````

**To modify a variable from the outer scope, declare it with nonlocal (or global, if needed).**

````python
def outer_function():
    count = 0               # variable of outer function

    def inner_function():
        nonlocal count      # access to outer function variable
        count += 1
        return count

    return inner_function

counter = outer_function()
print(counter())  # Output: 1
print(counter())  # Output: 2
````
