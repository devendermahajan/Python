## **Exception Handling**
-----
Python program may generates unexpected output or behaves abnormally. These are known as *bugs* or *errors*.

An *error* is a general term used to denote an unexpected scenario in hardware or software, which causes it to malfunction. *If errors are found in the source code, interpreter stops the execution at the line where error was detected*.

### **Type of Errors**

1. SYNTAX ERROR: 

    These errors usually occur when one or more grammatical rules of the programming language being used is violated. Such errors are detected during compile time. These errors are also known as *parsing errors* or *compile-time errors*. On encountering a syntax error, the interpreter does not execute the program unless we rectify the errors, save and rerun the program. 

````python
message = "Hello World!"        # Okay
````

````python
if message=="Hello"
    print("Okay")
````

2. LOGICAL ERROR:  

    A logic error is a problem in the program that causes it to produce an incorrect result. As interpreter or compiler does not report these types of programming errors, they are the most difficult to detect. Unlike syntax or runtime errors, *logical errors can be challenging to detect and fix because the code runs without producing any error messages*.

````python
# n! = n * (n-1) * (n-2) * ... * 1
# 5! = 5*4*3*2*1 = 120
n=5
f = 1
for x in range(1,n):
    f *= x
print(f"{n}! = {f}")
````

3. RUNTIME ERROR:  

Program terminates while executing due to wrong inputs. These errors are displayed instantly during the program execution. A runtime error stops the program execution, and a diagnostic message is displayed. It is *trappable error*. *Ex: Dividing a number by zero*. These are known as `Exceptions` in Python. 

````python
# Okay
dividend = 45
divisor = 5
print(dividend/divisor)
````

## **WHAT IS AN EXCEPTION?**

*An exception is a Python object that represents an error*. When an error occurs during the execution of a program, an exception is raised. 

A programmer required to anticipate such erroneous situations that may arise during its execution and handle them by including appropriate code to nuteralize that exception.

*Some common examples of Exceptions are* :
-	Divide by zero errors
-	Accessing the elements of an array beyond its range
-	Invalid input
-	Hard disk crash
-	Opening a non-existent file
-	Heap memory exhausted

#### **Advantages of exception handling are :**
```python

    # error prone code

[else:
    # error free code
finally:
    # cleaning code
]
```
1. Exception handling separates error-handling code from normal code.

````python
print("Welcome to Division Program!")
a = 10
b = 0
try:
    print (a/b)
except:
    print("Invalid operation.")

print("End of the program")
````

### **EXCEPTION HANDLING**
Exception Handling in Python involves the use of **`try`** and **`except`** clauses in the following format wherein the code that may generate an exception is written in the try block and the code for handling exception when the exception is raised, is written in except block.

**Syntax:**
```py
try:
    # code that may generate an exception
except:
    # what to do when the exception has occurred
```

````python
a = 22
b = 0
print(a/b)
````

````python
# Managing ZeroDivisionError with `if`
a = 22
b = '0'
print(a/b if b!=0 else "Division by zero is not possible.")
````

````python
a = 22
b = '0'
res=None
if not isinstance(b,(int,float)) or not isinstance(a,(int,float)):
    res=f"Division between {type(a)} and {type(b)} is not possible"
elif b==0:
    res=f"{a}/{b} is not possible"
elif b!=0:
    res = f"{a}/{b} = {a/b}"
print(res)
````

> In following snippet, all exceptions will be handled by the single `except` block.

````python
a = 22
b = 1
try:
    # code segment that may raise an exception
    c = a/b
    print(c)
except:        # will catch all exception 
    print("Error occured while division")
````

#### **Exception Type**
> Specific exception type can be used for stricter exception handling.

#### **ARGUMENT OF THE EXCEPT BLOCK**
In previous example, no argument has been specified. Python evaluate the error and raise exception accordingly. But programmer can also provide a second argument for the `except` block, which gives a reference to the exception object.

**Syntax**:
```py
try:
    # code that may generate an exception
except <exception-name> [as <argument>]:
    # what to do when the exception has occurred
```
The `except` clause can then use this additional argument to print the associated error-message of this exception as: `str (argument)`. 

````python
a = 22
b = 0
try:
    c = a/b     # code segment that may raise an exception
    print(c)
except ZeroDivisionError:        # only catches division by zero
    print("Error: Division by zero is not possible.")
````

### **Multiple Exceptions**
Multiple exception handling can be useful to provide a more exact error message to the user than a simple error message. In order to capture and process different type of exceptions, there must be multiple exception blocks – each one pertaining to different type of exception.

**Syntax**:
```py
try:
    # code that may generate an exception
except <exception-name1> [as <argument>]:
    # what to do when the exception1 has occurred
except <exception-name2> [as <argument>]:
    # what to do when the exception2 has occurred
…
else:
    # what to do when no exception has occured
```
The `else` clause executes code if there is no exception raised.


````python
# Multiple exceptions
a = 22
b = 0
try:
    c = a/b     # code segment that may raise an exception
except ZeroDivisionError:        # will catch all exception 
    print("Error: Division by zero is not possible.")
except TypeError:
    print("Error: Division is only possible between numeric values.")
else:
    print(c)
````

````python
# Default message of exception
a = 22
b = 0
try:
    c = a/b     # code segment that may raise an exception
    print(c)
except ZeroDivisionError as e:        # will give access to default error message
    print("Error:", e)
except TypeError:
    print("Error: Division is only possible between numeric values.")
````

#### **THE FINALLY BLOCK**
It defines a block of code to run when the `try... except...else` block is final. The code in `finally` block will be executed no matter if the `try` block raises an error or not. This can be useful to close objects and clean up resources.

**Syntax:**
```py
try:
    # code that may generate an exception
except <exception-name> [as <argument>]:
    # what to do when the exception has occurred
[else:
    # what to do when no exception has occurred]
finally:
    # statements that will always execute
```

````python
# Multiple exceptions
a = 22
b = '0'
c = None
try:
    c = a/b     # code segment that may raise an exception
except ZeroDivisionError:        # will catch all exception 
    print("Error: Division by zero is not possible.")
except TypeError as e:
    print(str(e))
else:
    print(c)
finally:
    del a
    del b
    del c
    print("Resources free. application closed")
````

### **RAISING AN EXCEPTION**
In Python, the `raise` keyword can be used to force an exception. That means, programmer can force an exception to occur through `raise` keyword. It can also pass a custom message to exception handling module. 

**Syntax:**
```py 
raise <exception> (<message>)
```

````python
# Multiple exceptions
a = 22
b = 0
c = None
try:
    if not isinstance(a,(int,float)) or not isinstance(b,(int,float)):
        raise TypeError (f"Division between {type(a)} and {type(b)} is not possible.")
    if b == 0:
        raise ZeroDivisionError(f"{a}/{b} can't be determined.")
except ZeroDivisionError as e:        # will catch all exception 
    print(e)
except TypeError as e:
    print(e)
else:
    c = a/b
    print(c)
finally:
    del a
    del b
    del c
    print("Resources free. application closed")
````

### **THE ASSERT STATEMENT**
An `assert` statement in Python is used to test an expression in the program code. 
- If the result after testing comes false, then the exception is raised, 
- otherwise nothing is done. 

This statement is generally used in the beginning of the function or after a function call to check for valid input. 

**Syntax:** 
```py
assert expression [,error_message]
```

On encountering an `assert` statement, Python evaluates the expression given immediately after the `assert` keyword. If this expression is `false`, an `AssertionError` exception is raised. 

````python
a = 10
b = 0
assert b!=0, "Division Error"       # 2 != 0 T # 0!=0 F
print(a/b)
````
