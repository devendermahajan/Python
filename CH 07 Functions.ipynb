## **CHAPTER 7: FUNCTION**
----
Functions are one of the fundamental building blocks in Python programming language. Function can be defined as a group of logically related statements that performs a particular task.


### **FEATURES OF A FUNCTION** 

- **Modular Programming:** Functions, divide a big program into small parts in such a way that each part performs some specific tasks. 

- **Reusability of Code:** Once, a function is defined, it can be called many times. This feature saves the programmer from rewriting the same code again.

- **Manageability:** Functions make the program manageable. It becomes easy to locate bugs and therefore make the program reliable. It also becomes easy to carry out local optimization in a function. 

    - **Easy Debugging:** Once bugs are located, it becomes very easy to rectify them. It becomes easy to effectively analyse smaller functions rather than the whole task.

    - **Efficient:** It is essential to make code efficient both in terms of time and memory. With functions, the code optimization is carried out more effectively and efficiently.

### ***PART OF FUNCTION***
- **Function Name:** A function name follows the same rules as other identifiers follow. It can have any legal literal name. For example, `sumofDigits` is a valid function name whereas `2strong` is not. 

- **Arguments:** The arguments of a function denote the input given to a function. A function can have zero or any number of arguments. These are also known as *parameters*.

- **Return Value:** A function may or may not return a value. The elegancy of Python lies in not specifying the return type and therefore using the same functions for various data types.

### **TYPE OF FUNCTIONS**
In Python, functions can be broadly classified as:

1.	Built-in Functions
2.	User defined Functions.


#### **Built-in Functions**
Python has several functions that are readily available for use. These functions are called built-in functions. Python Built-in functions are standard library functions that are part of language itself.

1. `abs (value)` function returns the absolute value of a number. A negative value’s absolute is that value is positive

```python
num = 7
print("Absolute value of",num,"=",abs(num))
num = -7
print("Absolute value of",num,"=",abs(num))
```

2.	`all (iterable)` function returns `True` if all values in terable like list, dictionary, set, etc., are of a boolean value of `True`, otherwise `False`

```python
l1 = [1,2,3,4]
print("all(l1):",all(l1))
l1 = [1,2,0,4]
print("all(l1):",all(l1))
```

3.	`any (iterable)` function returns `False` if all values in iterable like list, dictionary, set, etc., are of a boolean value of `False`, otherwise `True`

```python
l1 = [1,2,0,4]
print("any(l1):",any(l1))
l1 = [None,False,0]
print("any(l1):",any(l1))
```

4.	`bin (integer)` converts an integer to a binary string. But it can’t be applied on floats.
5.	`hex (integer)` converts an integer to its hexadecimal representation. But it can’t be applied on floats.
6.	`oct (integer)` converts an integer to its octal representation. But it can’t be applied on floats.

```python
num = 543
print("Binary of",num,"=",bin(num))
print("Octal of",num,"=",oct(num))
print("Hexadecimal of",num,"=",hex(num))
```

7.	`bool (value)` converts a value to Boolean. Every value except 'None',0, empty iterator is converted to `True`
8.	`int (value)` function converts a float or a string into an integer. In case of string, for ill-formatted string, function raises ValueError exception. 
9.	`float (value)` function converts an integer or a string into a float. In case of string, for ill-formatted string, function raises ValueError exception. 


```python
num = '232'
print('Integer coversion of',num,'=',int(num))
print('Float coversion of',num,'=',float(num))
print('Boolean coversion of',num,'=',bool(num))
num = None
print('Boolean coversion of',num,'=',bool(num))
```

10.	`callable (object)` returns `True`, if the object appears to be callable, otherwise returns `False`.	

```python
print("Is",num,'callable?',callable(num))
prn = lambda x: x**2
print("Is",prn,'callable?',callable(prn))
print("Square of 16 =",prn(16))
```
