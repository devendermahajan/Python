## **USER DEFINED FUNCTION DEFINITION**

A user defined function is a block of program statements which can be used repetitively in a program. It saves the time of a developer. In Python concept of function is same as in other languages.

### **ADVANTAGES OF USER-DEFINED FUNCTIONS**
- User-defined functions help to decompose a large program into small segments which makes program easy to understand, maintain and debug.

- If repeated code occurs in a program. Function can be used to include those codes and execute when needed by calling that function.

- Programmers working on large project can divide the workload by making different functions.

## **FUNCTION DEFINITION**

The definition of a function specifies the behaviour of the function. In order words, the task to be performed by the function is contained in the definition. A function can be create using following syntax: 

```python
def function-name ([parameters]): 
    # function body
```

- **def**: indicates that function definition is starting.

- **function-name**: Function name can be any valid identifier except Python keywords. 

- **parameters:** They are a kind of inputs to the function.The parameter list is given in the round parentheses () 
    - If function has no parameters, then () should be left empty. 
    - If function has parameters, each parameter should be each separated by a comma (,). 
    
    Parameters can be classified as formal parameters and actual parameters.
    - *Formal Parameters*: Parameter names given in the function definition.
    - *Actual Parameters*: Values/variables that are passed into the function during call.

- **Body of the function**: The body of the function contains the code that implements the task to be performed by the function. If UDF is returning some value back to the calling statement, it must have a `return` statement.

### **FUNCTION INVOCATION** 
The invocation of a function refers to calling a function. The invocation of a function can be at any place after the definition. A function can be called any number of times after its legal definition.

````python
# simple interest function
# si = P * R * T / 100
def SimpleInterest(Principal,InterestRate,Period):
    intr = Principal * InterestRate * Period / 100
    print(round(intr,2))
````

````python
SimpleInterest(1500,7.5,5)
````

### **TYPES OF UDF**
Based on the parameters and the return type, functions can be divided into the following categories.
1. The first type of function does not take any parameter or return anything.
2. The second type of function takes parameters but does not return anything.
3. The third type of function takes parameters and returns something back.

````python
# 1. The first type of function does not take any parameter or return anything.
def GoodMorning():
    print("Good Morning!")
````

````python
GoodMorning()
GoodMorning()
GoodMorning()
````

````python
# 2. The second type of function takes parameters but does not return anything.
# all parameter are mandatory and are handled sequentially
def Greetings(name,message):
    print(f"{message} {name}! ")
````

````python
Greetings("Ram","Good Morning")
Greetings("Shyam","Good Evening")
````

````python
# all parameters are mandatory
Greetings()
````

````python
# handled sequentially: first value in function is mapped to first parameter in () and so on.
Greetings("Have a nice day",'Mohan')
````

````python
# 3. The third type of function takes parameters and returns something back.
def product(a,b):
    return a*b
````

````python
# passing two numbers
print(product(4,5))
````

````python
# passing a list and number
print(product([3,4,5,3,2],3))
````

````python
# passing a number and string
print(product("*",50))
````

### **DEFAULT PARAMETERS**
Each default parameter has a default value as part of its function definition. Any calling function must provide arguments for all required parameters in the function definition but can omit arguments for default parameters. If no argument is sent for that parameter, the default value is used.

Usually, the default parameters are defined at the end of the parameter list, after any required parameters and non-default parameters cannot follow default parameters. The default value is evaluated only once.


````python
def SimpleInterest(Principal,InterestRate=1.0,Period=1):
    intr = Principal * InterestRate * Period / 100
    return(round(intr,2))
````

````python
# only principal amount given
SimpleInterest(199900)
````

````python
# principal amount and interest rate given
SimpleInterest(199900,4.5)
````

````python
# all parameters given
SimpleInterest(199900,4.5,6)
````

## **KEYWORD ARGUMENTS**
Whenever a function is called with values as its arguments, these values get assigned to the parameters in the function sequentially.
The order is especially important as values passed into these functions are assigned to corresponding parameters based on their position. And when the positions of the arguments are changed, the output produces different results.

````python
# parameter values are handled sequentially: first value in function is mapped to first parameter in () and so on.
Greetings("Have a nice day",'Mohan')
````

````python
# Keyword reposition.
Greetings(message="Have a nice day",name='Mohan')
````

## *ARGS AND **KWARGS
There are three types of arguments that a Python function can accept: 
- standard parameters: Standard arguments are the simplest but have limited functionality
- variable (*args), 
- keyword (**kwargs) make functions more flexible, by accepting a variable number of inputs. Here variable number of arguments means that the user does not know in advance about how many arguments will be passed to the calling function.

````python
# find average of 5 values
def stdAvg (a,b,c,d,e):
    return ((a+b+c+d+e)/5)

stdAvg(3,4,5,6,4)
````

````python
stdAvg(3,4,5,6)
````

````python
stdAvg(1,2,3,4,5,6,5)
````

````python
# using *args -> represents a tuple
def average(*vals):
    return sum(vals)/len(vals)
````

````python
average(3,4,5,6,4)
````

````python
average(3,4,5,6)
````

````python
import random as ra
l1 = [ra.randrange(1,21) for x in range(445)] 
print(l1)
````

````python
average(l1)
````

````python
# how to pass iterators like list or tuples
average(*l1)
````

````python
def SimpleInterest(Principal,InterestRate=1.0,Period=1):
    if not isinstance(Principal,(int,float)):
        return ("Error: Non numeric Principal amount passed.")
    if not isinstance(InterestRate,(int,float)):
        return ("Error: Non numeric rate of interest passed")
    if not isinstance(Period,int):
        return ("Error: Period should only be whole number.")
    intr = Principal * InterestRate * Period / 100
    return(round(intr,2))
````

````python
SimpleInterest(452025,5.65,6)
````

````python
SimpleInterest([452025],5.65,6)
````

## **LAMBDA FUNCTIONS** 
The `lambda` keyword is used to define anonymous functions in Python. Usually, such a function is meant for **one-time use**.

**Syntax:**
```python
func_obj = lambda [arguments] : expression
```

The `lambda` function can have zero or more arguments before the `:` symbol. When this function is called, the expression after : is executed.

````python
def double (x):
    return x*2
````

````python
double(9)
````

````python
dbl = lambda x : x*2
````

````python
dbl(9)  # 9*2
````
