## **Python**
------
Python is a high level, general purpose, strong procedural, object-oriented, and functional language crafted in late **1980**s by **Guido Van Rossum**. First official release was done in **1991**. It is popular due to its simplicity and robustness. It is easy to interface with other languages like, C++ and Java.

> Language was named in honour of British Comedy show *Monty Python’s Flying Circus*.

### **Python Features**
------
1.	The language and its source code are available to the public for free.
2.  Python is platform independent, it runs on Windows, Mac, Linux, or any other platform. 
3.  Python is very easy to learn and understand.
4. Python is more expressive than other languages due to being having simpler and shorter syntax
5.	It is interpreted(executed) line by line. 
6.	Python has support for an interactive mode, which allows interactive testing and debugging of snippets of code.
7.	It can be used to develop a GUI (Graphical User Interface). 
8.	Programmer can add low-level modules to the Python interpreter to add to or customize their tools to be more efficient.
9.	Python provides a better structure and support for large programs than shell scripting.
10.	Large Python Library: Python provides a large standard library.
11. Python is widely uesd in software development, web app development, desktop GUI development, education, and scientific applications.
12. No prior variable declaration is required. It supports automatic allocation and garbage collection.
13. It is also dynamically typed because it carries out type-checking at runtime.

### **POPULAR PYTHON IMPLEMENTATIONS**
-------
-	**CPython** - is the reference implementation, written in C.
-	**PyPy** - Written in a subset of Python language called RPython.
-	**Jython** - Written in Java.
-	**IronPython** - Written in C#.
All the implementations are compilers as well as interpreters. The compiler converts the Python program into intermediate bytecode. This bytecode is then interpreted by the interpreter.


### **PYTHON ARCHITECTURE**
1. **Parser:** It uses the source code to generate an abstract syntax tree.
2. **Compiler:** It turns the abstract syntax tree into Python bytecode.
3. **Interpreter:** It converts and executes the code line by line by line in a REPL (Read-Evaluate-Print-Loop) fashion. Stops on the line where error is found.

### **PYTHON WORKING**
- **Step 1.**	The programmer edits the source code using any code editor and save the source code to a file with `.py` extension.
- **Step 2.**	The source code is compiled by Python interpreter into bytecode.
- **Step 3.**	The bytecode is translated by the Python virtual machine into instruction that can interact with the operating system of the computer.


### PYTHON CODING CONVENTIONS
------
-	Each statement is terminated by pressing Enter key after a physical code-line.
-	Block of codes are denoted by line indentation (4 spaces not TABS) per indentation level.
-	Although multiple statements on one line are discouraged and not recommended in Python, but multiple statements can be combined using a semicolon (;) between two statements.
    ```python
    >>> a = 10
    >>> b = 20
    ```
    can be also written as 
    ```python
    >>> a = 10; b = 20
    ```
-	Python is case-sensitive language and most of its statement are written in lower case.
-	Docstrings are written using triple double quotes.


### WORKING MODE WITH A PYTHON 
-------
#### Interactive Mode

Python installation comes with an **Integrated Development and Learning Environment (IDLE)**. The best place to experiment with Python code and view the result instantly is in the interactive interpreter, otherwise known as a **shell**.

In this mode, the Interpreter executes the statement as soon as it is submitted to it and user get the result instantly. The Python shell is an excellent place to experiment with small code snippets.

The shell is a basically works on **R**ead-**E**val-**P**rint **L**oop (**REPL**) principles. 

##### How to use the interactive shell?
1.	Type the Python code after `>>>` prompt. Then press `Enter`.
2.	For valid code, shell displays the results. 
3.	For invalid code, shell displays relevant error message in <span style="color:red">**red**</span> colour. *The last line of Traceback dump explains the exact problem with the code*.

#### SCRIPT MODE 

In this mode, statements are written in separate file and is saved with extension `.py`. The interpreter reads the statements in the file executes them line by line statement as soon as it is submitted to it and user get the result instantly.
