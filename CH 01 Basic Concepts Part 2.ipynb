### CHARACTER SET
---------
The Python character set is a valid set of characters recognized by the Python language. These are the characters are used during writing a script in Python. Python supports all EBCDIC (IBM), ASCII (PC Compatibles) and Unicode characters that include:
* Alphabets: (`A-Z`) and (`a-z`).
* Digits: All digits `0-9`.
* Special Symbols: 	`" ' l ; : ! ~ @ # $ % ^  & * ( ) _ + – = { } [ ] \` .	
* White Spaces: tab space, blank space, newline, and carriage return.
* Other: All ASCII and UNICODE characters.


### KEYWORDS 
---------
These are reserved words that have a special meaning to the language translator and represent some definitive functionality. These words cannot be used as identifier names. There are `35` keywords in Python.

`True, False, None, and, or, not, is, if, elif, else, while, for, break, continue,` <br>
`return, in, yield, try, except, finally, raise, assert, import, from, as, class,` <br>
`def, pass, global, nonlocal, lambda, del, with` 

Except `True`, `False`, & `None` reserved words, all contain only lowercase alphabet symbols.


### Examples of Python keywords include:
- `True`, `False`, `None`: Represent boolean values and the absence of a value.
- `and`, `or`, `not`: Logical operators.
- `if`, `elif`, `else`: Control flow statements for conditional execution.
- `for`, `while`, `break`, `continue`: Control flow statements for iteration and loop control.
- `def`, `class`, `lambda`: Keywords for defining functions, classes, and anonymous functions.
- `import`, `from`, `as`: Used for importing modules and aliasing.
- `try`, `except`, `finally`, `raise`, `assert`: Keywords for exception handling and debugging.
- `with`, `as`: Used for context management.
- `return`, `yield`: Used for returning values from functions and defining generators.
- `global`, `nonlocal`: Used for managing variable scope.
- `async`, `await`: Keywords for asynchronous programming.

### IDENTIFIERS
----
These are the basic building blocks. They represent the name given the different parts of the program such as variables, objects, classes, functions, etc. An identifier starts with a letter `A` to `Z` or `a` to `z` or an underscore (`_`) followed by zero or more letters, underscores, and digits (`0` to `9`). 

### EXPRESSIONS
-----
An expression is a legal grouping of letters, symbols, and numbers being used to represent the value of one or more variables. 

**Example:**
```python
a = 10
b = 20
c = a + b
```

> In the expression `c = a + b`, `a`, and `b` are operands while `+` is the operator.

### STATEMENTS
-----
A statement is a single line of code written legally in a programming language that expresses an action to be carried out. 
* A statement might have internal components of its own, including expressions, operators, and functions. 
* A statement may or may not yield a value on execution. 
* A program is nothing but a sequence of one or multiple statements.

**Example:**
```python
>>> a = 10
>>> b = a - 5
>>> print(a, b)
    10 5
```

### COMMENTS
----
Comment is a block of text or statement that is deferred by the interpreter while execution.
-	This is a very effective method to add documentation to code.
-	It also restricts statement from execution.

Python provides two types of comments:
- **Single Line:** can be added to the code by putting a hash symbol (`#`) before the text and it ends with the end of the physical line. 
    -	The physical line beginning with `#` are the full line comments. 
    -	When `#` symbol is inserted anywhere instead of beginning of the line, it is an inline comment.

    ```python
    # This is an example of full line comment
    a = 10          # This is an example of an inline comment
    ```
- **Multiline Comment:** Although multiline comment is not supported in Python, they can be inserted into ways:
    - Adding `#` symbol in the beginning of every physical line in the multi-line comment. In simple words, include as many full line comments as required.
    - Type comment using pair of triple single/double quotes.
    ```python
    '''
    Long comments are useful for detailed.
    Information about the program working.
    '''
    ```
