## **CHAPTER: DICTIONARY IN PYTHON**

Dictionaries are mutable, unordered collections with elements in the form of a *`key:value`* pairs that associate *keys* to *values*. Dictionaries in Python have a "key" and a "value of that key". `Dictionaries are containers that associate keys to values`.

## **Features**
-	Dictionary elements are mutable, i.e., individual element can be changed using related key.
-	The keys and their associated values can be of different types.
-	Dictionary keys must be of immutable type and can be either a string or a number. 
-	Duplicate keys are not allowed in the dictionary. 
    In case of duplicate key, previously assigned value is replaced with new one.
-	Dictionaries remember the order in which the `key:value` pairs were inserted.
-	Dictionary values are accessible by their keys.
-	Internally dictionaries are indexed or arranged on keys.
-	Dictionary allow to create new member as well as remove existing one.
-	Dictionaries are also called *associative arrays* or *mapping* or *hashes*.


````python
# list structure
primes = [2,3,5,7,11,13,17,19]
print(primes[5])        # 6th element
````

````python
rec = ['Ram',23,True,'Una']
# name: 'Ram', age:23, admitted:True, city:'Una'
````

## **CREATING DICTIONARY**
### **DICTIONARY DISPLAY CONSTRUCT**
A dictionary can create following the given syntax. 

Syntax:     
```python
dobj = {key1:value1, key2:value2, ...}
```
-	the curly brackets {} mark the beginning and end of the dictionary,
-	each entry *key:value* consists of a pair separated by a colon (:).

````python
candidate = {'name': 'Ram', 'age':23, 'admitted':True, 'city':'Una'}
print(type(candidate))
print(candidate)
````

````python
import json
print(json.dumps(candidate,indent=2))
````

 The keys of a dictionary must be of immutable types, such as, a string, a number, or a tuple (containing only immutable entries). 
 Python will give <span style="color:red">TypeError:"unhashable type"</span> if keys are of mutable types.


````python
book = {'isbn':22231232321,'title':'Boring Lectures','topics':['how to sleep?','how to yawn?'],'pages':101}
print(json.dumps(book,indent=2))
````

````python
noerror = {1:'One',(2,3):'Two Three'}       # tuple as a key
print(noerror)
````

````python
error = {1:'One',[2,3]:'Two Three'}     # list as a key Error
print(error)
````

### **CREATING EMPTY DICTIONARY**

An empty dictionary can be created by initialize variable with empty curly braces `{}` or using `dict()` without any parameter.

Syntax:
```py
    dobj = {} | dict()
```

````python
# using {}
emd1 = {}
print(len(emd1))
````

````python
# using dict()
emd2 = dict()
print(len(emd2))
````

## **METHODS OF CREATING DICTIONARIES**
Apart from the methods shown in previous section, there are other methods for creating dictionaries. Let us explore them.

### **1. Adding key:value pair to an empty dictionary**
In this method, firstly an empty dictionary is created and then keys and values are added to it one pair at a time. 
1.	Create an empty dictionary, using empty curly braces `{}` or by using dictionary constructor `dict()`.
```py
    book = dict()
```
2.	Add key:value pairs, one at a time, as per following syntax:
```py
    book['isbn'] = 1234509876
```

````python
std1 = {}
print(std1)
````

````python
std1['id']=39322
std1['name']='Komal'
std1['class']='O-Level'
print(json.dumps(std1,indent=1))
````

````python
# read a line an count the words appear in it
text = "in this method firstly an empty dictionary is created and then keys and values are added to it one pair at a time"

freq = {}
words = sorted(list(set(text.split())))   # creates  set of unique words
for word in words:
    if word not in freq:        # search word as a key in dictionary
        freq[word] = text.count(word)

print(json.dumps(freq,indent=1))
````

### **2. Passing key:value pairs as keyword args to dict() constructor**
Here the argument names are taken as keys of String type and assigned data as value for keys

Syntax:
```py
    dobj = dict(arg1=value1, arg2=value2,...)
```
- argn will serve as key
- valn will serve as assoicated value to the key

````python
can = dict(id=10101,name='Gagan',std='Bcom')
print(can)
````

### **3.  Using zip() function**
 In this method, the keys and values are enclosed separately in parentheses as lists or tuples and are given as arguments to the `zip()`  function. which is then given as argument of `dict()`.

 ```python
 keys =[k1, k2, ..., kn]
 vals =[v1, v2, ..., vn]
 dobj = dict(zip(keys,vals))
 ```

````python
keys = [1,2,3,4]
vals =['One','Two','Three','Four','Five']
numname = dict(zip(keys,vals))
print(numname)
````

### **4. Passing list of tuples containing (key, value)**
In this method, one list or tuple argument is passed to `dict()`. This argument contains lists/tuples of individual `key:value` pairs. 

````python
data = [(1, 'One'), (2, 'Two'), (3, 'Three'), (4, 'Four')]
dobj = dict(data)
print(dobj)
````

## Collections available in Dictionary
- `keys()` collection of all keys defined in a dictionary
- `values()` collection of all assocaited values defined in a dictionary
- `items()` collection of all keys:value pair defined in a dictionary as a tuple element (key, value)

````python
print(book)
print("All keys:", book.keys())
print("All values:", list(book.values()))
print("All items:", list(book.items()))
````

## **TRAVERSING A DICTIONARY**
Traversal of a collection means accessing and processing each element of it. The for loop makes it easy to traverse or loop over the items in a dictionary.

**Syntax:**
```py
for item in dictionary:
    #to-do
```

By default, while transversing a dictionary, the keys are considered. Hence the loop variable `item` will be assigned the keys of `dictionary` one by one, which can be used inside the body of the for loop.


````python
# 1. Using dictionary keys collection
for key in book:        # book.keys()
    print(f"{key:>6}: {book[key]}")
````

````python
# 2. Using dictionary values collection
# No access to the keys
for value in book.values():
    print(value)
````

````python
# 3. Using dictionary items collection (key,value)
for key,value in book.items():
    print(f"{key:>6}: {value}")
````

## **DICTIONARY METHODS AND FUNCTIONS**

1. The `.clear()` method empties the referred dictionary of all key-value pairs
   **Syntax**: `dobj.clear()`

````python
candidate = {'name':'Ram', 'age':42, 'salary':45000}
print(candidate)
candidate.clear()
print(candidate)
````

2. The `.pop()` removes a key from a dictionary, if it is present, and returns its value. 

**Syntax:** `dobj.pop(key[, default])`

If `key` is present in dobj, .pop() removes `key` and returns its associated value and raises a `KeyError` exception if key is not found.

````python
candidate = {'name':'Ram', 'age':42, 'salary':45000}
print(candidate)
````

````python
print(candidate.pop('name'))
print(candidate)
````

````python
print(candidate.pop('name'))        # KeyError
````

````python
print(candidate.pop('name','Key not found'))
````

3. The `.popitem()` method removes the last *key-value* pair added from dictionary and returns it as a tuple. For empty dictionary, it raises a `KeyError` exception.

**Syntax:** `dobj.popitem()`

````python
candidate = {'name':'Ram', 'age':42, 'salary':45000}
print(candidate)
````

````python
item = candidate.popitem()
print ('Removed item:',item)
print(candidate)
````

4. The Python `.update(obj)` merges a dictionary with another dictionary or with an iterable of key-value pairs.

**Syntax:** `dobj.update(obj)`

If `obj` is a dictionary, `.update(obj)` merges the entries from `obj` into dictionary. 

- If the key is not present in dictionary, the key-value pair from `obj` is added.
- If the key is already present in dictionary, the corresponding value for that key is updated to the value from `obj`.

````python
candidate1 = {'name':'Ram', 'age':42, 'salary':45000}
candidate2 = {'city':'Una', 'salary':54000}
print(candidate1,candidate2,sep='\n')
````

````python
# update candidate1 from candidate2
candidate1.update(candidate2)
print(candidate1)
````

- The `obj` may also be a sequence of key-value pairs, similar to the `dict()` function is used to define a dictionary. For example, `obj` can be specified as a list of tuples.

````python
d1 = {'a': 1, 'b': 2, 'c':4}
print(d1)
````

````python
newdata = [('d',4),('c',3)]
d1.update(newdata)
print(d1)
````

- Alternatively, the values to merge can be specified as a list of keyword arguments

````python
d1 = {'a': 1, 'b': 2, 'c':4}
print(d1)
````

````python
d1.update(c=3,d=4,e=5,f=6)
print(d1)
````

### **DEL STATEMENT** 
The **`del`** keyword can be used to *in-place delete* an existing key in the dictionary and it raises an *KeyError* exception if the key is not found and hence non-existence of the key has to be handled. 

**Syntax:** `del dobj[key]`

- calling `del` upon dictionary object without key will remove the entire dictionary from the memory.

````python
candidate = {'name':'Ram', 'age':42, 'salary':45000}
print(candidate)
````

````python
del candidate['name']
print(candidate)
````

````python
del candidate['name']
print(candidate)
````

````python
del candidate
print(candidate)
````
