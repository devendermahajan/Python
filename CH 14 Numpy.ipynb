## **NUMPY VS LISTS**
The primary difference between a Python list and a NumPy array is that lists are general-purpose, flexible collections for mixed data, while NumPy arrays are specialized, high-performance structures for numerical computations.

### **Key Differences**
|Feature| Python List| NumPy Array|
|------|-----|------|
|**Data Types**|Heterogeneous - can store strings, ints, objects together.|Homogeneous - all elements must be the same type.|
|**Memory**|Stores pointers to objects; elements can be non-contiguous.| Stores raw data in a single contiguous block of memory.|
|**Operations**| Does not support direct mathematical operations, e.g., `list * 2` repeats the list.|  Supports element-wise "vectorized" operations, e.g., `array * 2` doubles every element.|
|**Performance**| Slower for mathematical tasks due to type-checking overhead.| Significantly faster (often 10–100x) for numerical data.|
|**Flexibility**| Highly flexible; easy to append/remove elements.|Fixed size once created; resizing requires creating a new array.|



````python
# Error due to incompatible data types in the list  
dl = [1,'2',3,4]
print(sum(dl))
````

````python
# Solution: Calculate the sum of only the numeric elements in the list
s = 0
for e in dl:
    if isinstance(e,(int,float)):
        s += e
print(s)
````

````python
# Element-wise addition of two lists is not possible using the + operator. It will concatenate the lists instead of adding the corresponding elements.
l1 = [1,2,3,4]
l2 = [4,5,3,4]
print(l1+l2)
````

````python
# Solution: produce a new list having the sum of items in two list at same index
l = len(l1)
res = []
for i in range(l):
    res.append(l1[i] + l2[i])

print(res)
````

````python
# Using NumPy for element-wise addition of two lists. 
import numpy as np
arr1 = np.array(l1)
arr2 = np.array(l2)
arr3 = arr1 + arr2
print(arr3)
````

## **NUMPY DATA TYPES**
NumPy supports a much greater variety of numerical types than Python does. 
- Integers: 
    - `int8`, `int16`, `int32`, `int64`: Signed integers of varying bit sizes. `int_` is a platform-dependent alias, typically int64 on modern systems.
    - `uint8`, `uint16`, `uint32`, `uint64`: Unsigned integers, also of varying bit sizes.
- Floats (Floating-Point Numbers): 
    - `float16`, `float32`, `float64`, `float128` Floating-point numbers with different levels of precision. `float_` is typically `float64`.
- Complex Numbers:
    - `complex64`, `complex128`: Complex numbers, represented by two floating-point numbers (real and imaginary parts). `complex_` is typically `complex128`.
- Booleans: `bool_` Boolean values (True or False).
- Other Data Types:
    - `datetime64`: For storing date and time information.
    - `timedelta64`: For storing time differences.
    - `str_` or `S`: For fixed-size strings.
    - `bytes_` or `B`: For fixed-size byte strings.
    - `void`: For raw data blocks.

````python
# without dtype specification, numpy will infer the data type from the list
my_array = np.array([ 10 , 12 , 14 , 16 , 20 , 25 ])
print (my_array)
print (my_array.dtype)
print (my_array.dtype.itemsize)
````

````python
# creating a datetime array by specifying the dtype as `datetime64`
# date format: yyyy-mm-dd
str_dates = np.array(["2025-10-04", "2026-05-06", "2024-11-04"])
print (str_dates)    # Output: ['2025-10-04' '2026-05-06' '2024-11-04']
print (str_dates.dtype)  # Output: <U10 (or similar, depending on the system)
print (str_dates.dtype.itemsize) # Output: 40 (or similar, depending on the system)
````

````python
# creating a datetime array by specifying the dtype as `datetime64`
# date format: yyyy-mm-dd
str_dates = np.array(["2025-10-04", "2026-05-06", "2024-11-04"], dtype=np.datetime64)
print (str_dates)    # Output: ['2025-10-04' '2026-05-06' '2024-11-04']
print (str_dates.dtype)  # Output: datetime64[D] 
print (str_dates.dtype.itemsize) # Output: 8 (or similar, depending on the system)
````

### **Changing Datatype**
The most effective way to change the data type of an existing NumPy array is by using the `astype()` method . This function *creates a new copy of the array* cast to the specified data type. 

To change the datatype, pass the desired data type as a string (e.g., 'int', 'float') or as a NumPy `dtype` object (e.g., `np.float64`) to the `astype()` method.

**Syntax:**
```python
new_array = exist_array.astype(new_datatype)
```

#### Converting float array to integer array

````python
# Original float array
arr = np.array([1.1, 2.8, 3.5])

print(arr)          # Output: [1.1 2.8 3.5]
print(arr.dtype)    # Output: float64 (or float32 depending on system)

# Convert to integer (truncates the decimal part)
int_arr = arr.astype('int') 

print(int_arr)          # Output: [1 2 3]
print(int_arr.dtype)    # Output: int64 (or int32 depending on system)
````

#### **Text array to datetime array**

````python
str_dates = np.array(["2025-10-04", "2026-05-06", "2024-11-04"])
print (str_dates)    # Output: ['2025-10-04' '2026-05-06' '2024-11-04']
print (str_dates.dtype)  # Output: <U10 (or similar, depending on the system)
print (str_dates.dtype.itemsize) # Output: 40 (or similar, depending on the system)
````

````python
dates = str_dates.astype('datetime64') # convert string array to datetime64 array
print(dates)         # Output: ["2025-10-04", "2026-05-06", "2024-11-04"] 
print(dates.dtype)   # Output: datetime64[D] (or similar, depending on the system)
print(dates.dtype.itemsize) # Output: 8 (or similar, depending on the system)
````

#### **Common Conversion Examples**
- Float to Integer: `arr.astype(int)` — Truncates decimal values toward zero.
- Integer to Float: `arr.astype(float)` — Often used to enable precise division or further numerical processing.
- Integer to Boolean: `arr.astype(bool)` — Converts 0 to `False` and any non-zero value to `True`.
- Strings to Numbers: `arr.astype(float)` — Useful if you've loaded numeric data that was originally formatted as strings.

## **ATTRIBUTES OF A NUMPY ARRAY**
NumPy array have various attributes that are useful in working with this array. Some important attributes of NumPy array are discussed below.  
- *`.ndim`*: ndim represents the number of dimensions (axes) of the ndarray.
- *`.shape`*: shape is a tuple of integers representing the size of the ndarray in each dimension. 
- *`.size`*: size is the total number of elements in the ndarray. It is equal to the product of elements of the shape.
- *`.dtype`*: dtype tells the data type of the elements of a NumPy array. In NumPy array, all the elements have the same data type.
- *`.itemsize`*: itemsize returns the size (in bytes) of each element of a NumPy array.


````python
# 2-d numpy array
''' 
[
[row0],
[row1],
[...],
[rown]
]
'''
row1 = [1,2,3,4,5]
row2 = [6,7,8,9,10]
row3 = [11,12,13,14,15]
narr = np.array([
    row1,
    row2,
    row3
], dtype=np.uint16) # assign unsigned int datatype

# printing the array
print(narr)

#.ndim represents the number of dimensions (axes) of the ndarray.
print(f"Axes in array: {narr.ndim}")

#.shape is a tuple of integers representing the size of the 
# ndarray in each dimension. 
print(f"Shape of array: {narr.shape}")

#.size: size is the total number of elements in the ndarray.
print(f"Total element in the array: {narr.size}")

#.dtype: dtype tells the data type of the elements of a NumPy
print(f"Datatype of items in array: {narr.dtype}")

#.itemsize: itemsize returns the size (in bytes) of 
#each element of a NumPy array.
print(f"Length of items in array: {narr.itemsize} byte(s)")
````

#### **np.array vs np.asarray**
|Feature| 	`np.array`|	`np.asarray`|
|------|------|------|

### **5. `np.linspace()` Method**

### **Transversing ndarray**
> It like walking through entire array using `for` loop such that individual element could be processed.

#### **1-D ndarray**
**Syntax:**
```python
    for value in ndarray:
        # statements
```

````python
# transverse of the array
arr = np.array([1, 2, 3, 4, 5, 6])
for value in arr:   # iterating over the first axis (rows)
    print(value)    # prints the entire row as a 1D array  
````

#### **2-D ndarray**
**Syntax:**
```python
    for row in ndarray:
        for value in row:
            # statements
```

````python
# transvers a 2D array
arr = np.array([[1, 2, 3], [4, 5, 6]])

for row in arr:     # iterating over the first axis (rows)
    for value in row:   # iterating over the second axis (columns)
        print(value, end=' ')    # prints the entire row as a 1D array
    print() # for new line after each row
````

## **NumPy Array Manipulation**
> Let us learn about some of the most common array manipulation functions in NumPy.

#### **Sorting NumPy Arrays**
The `np.sort()` function returns a sorted copy. 

**Syntax:**
```python
sorted_arr = np.sort(original_arr,descending=False)
```

- Numeric arrays are sorted by default in ascending order of numbers. 
- On the other hand, text arrays are sorted alphabetically.

#### **Sorting 1-D Numpy Array**

````python
# sorting a numeric array
arr = np.array([5, 2, 9, 1, 5, 6])
print("Original Array:", arr)
sorted_arr = np.sort(arr)
print("Sorted Array:", sorted_arr)
````

````python
# sorting a text array
text_arr = np.array(["banana", "apple", "cherry", "date"])  
print("Original Text Array:", text_arr)
sorted_text_arr = np.sort(text_arr,descending=True)
print("Sorted Text Array:", sorted_text_arr)
````

#### **Sorting 2-D Numpy Array**

**Syntax:**
```python
sorted_arr = np.sort(original_arr,axis = {0|1} descending=False)
```

````python
# sorting a 2D array by row
arr_2d = np.array([[3, 5, 2, 4],
                   [7, 2, 6, 8],
                   [1, 4, 3, 6]])
print("Original 2D Array:")
print(arr_2d)
sorted_arr_2d = np.sort(arr_2d, axis=1)  # sort each row
print("2D Array Sorted by Row:")
print(sorted_arr_2d)
````

````python
# sorting a 2D array by column
arr_2d = np.array([[3, 5, 2, 4],
                   [7, 2, 6, 8],
                   [1, 4, 3, 6]])
print("Original 2D Array:")
print(arr_2d)
sorted_arr_2d = np.sort(arr_2d, axis=0)  # sort each column
print("2D Array Sorted by Column:")
print(sorted_arr_2d)
````

### **Broadcasting NumPy Arrays**

## **Statistical Operations with NumPy**
NumPy arrays contain various functionalities for performing statistical operations, such as finding the mean, median, and standard deviations of items in a NumPy array.

### `.mean()` Method
The `.mean()` method returns the average of all items in a Numpy array.

**Syntax:**
```python
var = np.mean(array)
```

````python
# Finding mean of an 1-D array
arr = np.array([12, 96, 78, 74, 55])
mean = np.mean(arr)
print("Mean of the array:", round(mean,2))
````

````python
# Finding mean of a 2-D array
arr = np.array([[12, 22, 43], [74, 55, 66]])
print("Original Array:")
print(arr)

# mean of all elements in the array
mean = np.mean(arr) 
print("Mean of the array:", round(mean,2))

# mean of each column
col_mean = np.mean(arr, axis=0)
print("Mean of each column:", np.round(col_mean,2)) 

# mean of each row
row_mean = np.mean(arr, axis=1)
print("Mean of each row:", np.round(row_mean,2))
````

### **Standard Deviation**
The standard deviation of a NumPy array can be found via the `.std()` method.

> **Standard deviation** is a statistical measurement that shows how far individual data points are spread out from the average (mean). A low standard deviation means data is clustered closely around the mean, while a high standard deviation indicates a wider spread

````python
# Finding standard deviation of an 1-D array
arr = np.array([12, 22, 43, 74, 55])
print("Mean of the array:", round(np.mean(arr),2))
std_dev = np.std(arr)
print("Standard Deviation of the array:", round(std_dev,2))
````

It indicates how variables move together—whether they increase, decrease, or stay the same in tandem—without proving that one variable causes the other. The strength is measured by a coefficient, usually, ranging from -1.0 to +1.0.

## **Array Comparisons**

### **`.where() method`**
The `numpy.where()` is used for conditional selection and replacement in NumPy arrays. It can be used to:
- Find indices that satisfy a condition
- Build a new array by choosing values from two options depending on a condition.

**Syntax:** 
```python
np.where(condition [, truearray][, falsearray])
```

> `np.where(...)` with a single argument returns a tuple containing the indices of `True` entries, as show in following example:

````python
# find the indices of elements in an array that satisfy a condition
arr = np.array([1, 2, 3, 4, 5])
indices = np.where(arr > 3)
print("Indices of elements greater than 3:", indices)
````

> By providing `truearray` and `falsearray` as arguments, `numpy.where()` return different values depending on whether condition is `True` or `False`.

````python
# use np.where() to check if elements in an array are even or odd
arr = np.array([1, 2, 3, 4, 5])
result = np.where(arr % 2 == 0, "Even", "Odd")
print("Even or Odd:", result)
````

#### **Picking from two arrays elementwise**
For elements where condition is `True`, corresponding element from `truearray` is chosen. Otherwise, element from `falsearray` is selected.

````python
# using where pick elements from two arrays based on a condition
arr1 = np.array([10, 25, 15, 30, 20])
arr2 = np.array([100, 200, 150, 300, 250])
# if arr1 > 20, pick element from arr1, otherwise pick element from arr2
result = np.where(arr1 > 20, arr1, arr2)  
print("Result of np.where() with two arrays:", result)  
````

#### **Replace negatives by zero**

````python
# replace negative values in a 2d array with 0
arr = np.array([[-1, 2, -3], [4, -5, 6]])
print("Original Array:")
print(arr)
arr = np.where(arr < 0, 0, arr)
print("Array with negative values replaced:\n", arr)
````

## **Boolean Indexing**
Boolean indexing in NumPy is a method for selecting elements from an array based on conditions rather than their integer positions. This technique uses a boolean mask—a binary array of `True` and `False` values—to filter data.

**Syntax:**
```python
new_array = existing_array[condition]
```

````python
# boolean indexing to filter elements in an array
arr = np.array([1, 2, 3, 4, 5])
filtered_arr = arr[arr > 3]
print("Filtered Array (elements greater than 3):", filtered_arr)
````

#### **Features**
- Unlike basic slicing which returns a "view," boolean indexing always returns a new copy of the data.
- It is primarily used to extract subsets of data that meet specific criteria.

### **In-place Modification**
Boolean masks can be used to modify specific elements directly. For example, following snippet will set all negative values to zero.

````python
# In place modification of an array using boolean indexing
arr = np.array([1, -2, -3, 4, 5]) 
arr[arr < 0] = 0  # set negative numbers to 0
print("Array with negative values replaced:", arr)
````

### **Combining Multiple Conditions**

Using bitwise logical operators one can create complex filters by combining multiple boolean expressions.
- & (AND): Elements must satisfy both conditions.
- | (OR): Elements must satisfy at least one condition.
- ~ (NOT): Inverts the condition. 

````python
# using bitwise operators selects values between 10 and 40
arr = np.array([5, 15, 25, 35, 45])
result = arr[(arr >= 10) & (arr <= 40)]
print("Filtered Array (values between 10 and 40):", result)
````
