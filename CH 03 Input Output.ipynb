## **INPUT/OUTPUT IN PYTHON**
---

## **DEVICES**
1.	**Standard Input**: Default source from where data shall be read, i.e., keyboard.
2.	**Standard Output:** Default destination where result shall be sent, i.e., monitor.
3.	**Standard Error:** Default destination where errors shall be sent, i.e., monitor.


### READING INTO VARIABLES.
In Python, `input ()` function is used to gather data from the user. The syntax for `input ()` function is,
```python
var=input (["prompt"]) 
```
- The _prompt_ (optional) is the string that is printed on the string which gives an indication to the user of the value that needs to be entered. 

````python
age = input("How old are you? ")
print(age)
````

- When the user presses Enter key, the program resumes, and `text value is returned by input function, even if digits has been entered`.

````python
print(type(age))
````

### READING NUMERIC DATA

To cast input data into numeric values while reading, ``input()`` function can be used in given syntax:
```python
var=eval|int|float(input(["prompt"])) 
```

- `eval ()` converts data into appropriate format automatically or raise Error.
- `int ()` converts data into integer or raise Error.
- `float ()` converts data into floating number or raise Error.


````python
"""
input function accepts a string value from keyboard and pass it to `int` function which returns an integer value after converting string into integer.
int function will raise ValueError Exception of invalid numeric format if string contains characters other than digits.
"""
age = int(input("How old are you? "))
print(age, "is of type", type(age))
````

````python
"""
input function accepts a string value from keyboard and pass it to `float` function which returns a standard fractional value after converting string into float.
float function will raise ValueError Exception of invalid numeric format if string contains other characters than digits and a decimal point.
"""
age = float(input("How old are you? "))
print(age, "is of type", type(age))
````

````python
"""
input function accepts a string value from keyboard and pass it to `eval` function which returns an integer or float value after converting string.
eval function will raise ValueError Exception of invalid numeric format
"""
age = eval(input("How old are you? "))
print(age, "is of type", type(age))
````

## **DISPLAYING VARIABLES OR TEXT**
-----
The `print()` function is used to provide output to the user on the monitor. 
- By default, `print` statement will displays output on separate line for each `print` function. 
- The `print` statement will print everything as strings and non-string is automatically converted to its string representation.

### **DISPLAY A MESSAGE**
To display a message, pass string literal enclosed in the pair of either single quotes or double quotes in the `print` statement, as shown:

```python
print ("message")
```

````python
print("Hello! Welcome to Python Programming")
print('Have a nice day!')
````

### **DISPLAY VALUE OF VARIABLE**
To display the value of a variable, pass the variable name in the `print` statement.

``` py
print (var-name)
```

````python
prime = 17
print(prime)
````

### **DISPLAY MULITPLE VALUES**
To display multiple values, pass each value separated by a comma (,) in the `print` statement. Each output parameter can be of different type. 
> Note that `print` statement will insert a space after each output parameter.
``` py
print (out1, out2, …)
```

````python
# I am 52 years old
age = 52
print("I am",age,"years old.")
````

### **FEATURES OF PRINT STATEMENT**
-	It automatically converts the non-textual data to string before output.

````python
print("I am",age,"years old.")      # age is of type integer
````

-	It inserts a *space* between each item separated by a comma in `print` statement. It uses `sep` argument to specify the separator character  which is space by default. It can be changed to get different result.
    ``` py
    print (*outputs [,sep=' '])
    ```

````python
a = 10
b = 20
print(a,b)          # outputs separated by space
````

In the example below, value of `sep` argument is set to colon (`:`).

````python
print(a,b,sep=":")  # outputs separated by colon
````

````python
# print each argument on separate line using sep parameter
print("Single statement - output on separate lines")
print(a,b,sep='\n')     # outputs separated with a new line (\n)
````

-	It automatically appends a newline character (`\n`) unless the default value of `end` argument is not changed.

    ```py
    print (*outputs [,end='\n'])
    ```

````python
s1 = 'Python is an easy language'
s2 = 'we should learn it'
# Each output will be printed on separate lines
print (s1)
print (s2)
````

````python
print(s1,s2)
````

````python
# setting 'end' argument to 'and'
print(s1,end=' and ')
# output of next print statement will be printed on previous line
print(s2)
````

````python
a=10
b=2
c=a*b
# 10 x 2 = 20
print(a,'x',b,"=",c)
````

### **F-STRINGS**
> Formatted strings or *f-strings* were introduced in `Python 3.6A`, *f-string* is a string literal that is prefixed with `f`. These strings may contain replacement fields, which are expressions enclosed within curly braces `{}`. The expressions are replaced with their values.

``` py
f"message {expr[:[width][.precision][type]]}"
```
where
- *width* specify total space to be assigned to the `expr`.
- *precision* specify no. of digits should appear after decimal point.
- *type* specify how output to be formatted.

````python
a=10
b=2
c=a*b
# 10 x 2 = 20
print(a,'x',b,"=",c)        ## std method: each output param is specified separately
# using f-string
print(f"{a} x {b} = {c}")
````

````python
num=34.35723424883
print("Value =",num)
````

````python
# only two decimal digits
print("Value =",round(num,2))
````

````python
# {expr:.2f} -> Only 2 digits after decimal
print(f"Value = {num}")
print(f"Value = {num:.2f}")
````

#### **TYPE CHARACTER**


| Sym | Description |
| --- | ----------- |
|d| Formats the output as **integer** when used inside the placeholder.|
|b| Formats the output in **binary** format when used inside the placeholder.|
|o| Formats output in **octal** number.|
|x or X| Formats output as **hex** number in lowercase (`x`) or uppercase (`X`)|
|f| Formats the output in a **fixed-point** number format. By default, it gives the output with six decimal places. To get up to 2 decimal places, use it as `.2f`.|
|e or E| Formats the output in **scientific** format. The exponent will be in lowercase for `e` and in uppercase for `E`.|
|%| Formats the output in a **percentage** format (6 decimal places) To truncate any decimal value use :.0%.|
|, or _| Sets **comma** (,) or  an **underscore** (_) as a thousand separator.|
|c| Adds spaces before any positive numbers.| 
|<| Left align the result. |
|^| Centre align the result.| 
|>| Right-align the result. | 

````python
# formatting integer values
bin_var = 0b1110    # binary
print(f"Integer value of binary {bin_var:b} is {bin_var}")
int_var = 125         # integer
print(f"Binary value of integer {int_var} is {int_var:b}") 
print(f"Octal value of integer {int_var} is {int_var:o}")
print(f"Hexadecimal value of integer {int_var} is {int_var:x}")
````

````python
# formatting float values
float_var = 123.452789567667656809
print(float_var)                            # normal precision
print(f"Float value is {float_var:f}")      # default is 6 decimal places
print(f"Float value is {float_var:.2f}")    # 2 decimal places
print(f"Float value is {float_var:.0f}")    # without decimal places
print(f"Float value is {float_var:E}")      # scientific notation
print(f"Float value is {float_var:.2e}")    # scientific notation with 2 decimal places
````

````python
# formatting as % value
percent_var = .753
print(f"Percentage value is {percent_var:%}") # default 6 digits places
print(f"Percentage value is {percent_var:.0%}") # 0 decimal places 
print(f"Percentage value is {percent_var:.1%}") # 1 decimal places
print(f"Percentage value is {percent_var:.2%}") # 2 decimal places
````

````python
# formatting number with comma separator
num = 1234567890    # integer   
print(f"Number with comma separator: {num:,}")          # comma separator
print(f"Number with underscore separator: {num:_}")     # underscore separator

newnum=1_234_434_546_544_344                            # using  
print(newnum)
````

````python
# formatting string
name = "Ram"
int_var=123
print(f"{name} Himachal")      # simple string formatting
print(f"{name:15}Himachal")   # left aligned  | total width 15
print(f"{name:>15}")  # right aligned  | total width 15
print(f"{name:^15}")  # center aligned  | total width 15
print(f"{name:-^15}") # center aligned  | total width 15 | fill character *  | 
````

````python
# formatting string
int_var=123
print(f"Integer value: {int_var}")
print(f"Integer value: {int_var:10}")   # default alignment is right for numbers
print(f"Integer value: {int_var:<10}")
print(f"Integer value: {int_var:*>10}")   # default alignment is right for numbers
````
