## Operators

2.  Assignment Operators
3.  Relational Operators

5.  Logical Operators

### Arithmetic Operators
Each of these operators is a binary operator i.e., it requires two values (operands) to calculate a final answer. Apart from these binary operators, Python provides two unary arithmetic operators (that require one operand) also, which are unary $ + $, and unary $ - $.

For `a=10` and `b=4`

| Operator | Description |Expression|Result|
| --- |  --- | --- | --- |
| /  | Divides its first operand by the second operand and always result as a float value. | `c = a/b` | `2.5` |
| // | Divides its first operand by the second operand and returns integral part, i.e., quotient of the division.  | `c = a//b` | `2` |
| %  | Produces the modulus i.e., remainder of dividing the first operand by the second operand.  | `c = a%b` | `2` |
| ** | Returns the result of a number raised to a power. | `c = a**b` | `10000` |
|  | Use `0.5`or `(1/2)` as power to calculate square root of a number. | `c = 4**0.5` | `2.0` |
|  | Use `(1/3)` as power to calculate cube root of a number. | `c = 27**(1/3)` | `3.0` |
| *  | Returns the product of its operands. | `c = a*b` | `40` |
| +  | Adds values of its numeric operands and the result is the sum values of its two or more operands. Incompatible data values will throw an exception. | `c = a+b` | `14` |
| -  | Subtracts the second numeric operand from the first. | `c = a-b` | `6` |

````python
a = 13      # variable a created with value 14
b = 4       # variable b created with value 4
print(a,b)
````

````python
# float division
c = a / b
print(a,'/',b,'=',c)
print(f'{a} / {b} = {c}')
````

````python
# floor division (Quotient)
c = a // b
print(f"{a} // {b} = {c}")
````

````python
# modulo division (Remainder)
c = a % b
print(f"{a} % {b} = {c}")
````

````python
# Raise to power
c = a ** b
print(f"{a} ** {b} = {c}")
print(f"Square root of 81 = {81**(1/2)}")       # Square root
print(f"Cube root of 27 = {27**(1/3)}")         # Cube root
````

````python
# product
c = a * b
print(f"{a} * {b} = {c}")
````

````python
# Addition
c = a + b
print(f"{a} + {b} = {c}")
````

````python
# difference
c = a - b
print(f"{a} - {b} = {c}")
````

### Assignment Operators

Assignment operators are used for assigning the values generated after evaluating the right operand to the left operand. Assignment operation always works from right to left. Assignment operators are either simple assignment operator or compound assignment operators.

#### Simple assignment Operator

> It is done with the equal sign (`=`) and simply assigns the value of its right operand to the variable on the left.

```python
x = 2
```
#### Compound assignment operators

> Compound assignment operators combine assignment operator with another operator with = being placed at the end of the original operator. It supports shorthand notation for avoiding the repetition of the left-side variable on the right side. These operators are also called _augmented assignment operators_.

| Operation | Alternative | Description |
| --- | --- | --- |
| a /= b | a = a / b | Value of b does float division to the value of a and result assigned to a. |
| a //= b | a = a // b | Value of b does floor division to the value of a and result assigned to a. |
| a **= b | a = a ** b | a<sup>b</sup> computed, and result assigned to a. |
| a *= b | a = a * b | Value of b multiplied to the value of a and result assigned to a. |
| a %= b | a = a % b | Value of b divides the value of a and remainder assigned to a. |
| a += b | a = a + b | Value of b added to the value of a and result assigned to a. |
| a -= b | a = a - b | Value of b subtracted from the value of a and result assigned to a. |

````python
a=10
b=4
a+=b    # a=a+b     # a = 10 + 4 = 14
print("a+=b : ",a)
a-=b    # a=a-b     # a = 14 - 4 = 10
print("a-=b : ",a)
a*=b    # a=a*b     # a = 10 * 4 = 40
print("a*=b : ",a)
a/=b    # a=a/b     # a = 40 / 4 = 10.0
print("a/=b : ",a)
a//=b   # a=a//b    # a = 10//4 = 2.0
print("a//=b : ",a)
a%=b    # a=a%b     # a = 2.0 % 4 = 2.0
print("a%=b : ",a)
````

### Relational Operators
These operators used compare values of operands and return either `True` or `False`. Expression creating using relational operator are known as **relational expression**. Relational operators work with all datatypes according to following principles:

*   Numeric values are compared after removing trailing zeros after decimal point from a floating-point number. So,`4` and `4.0` will be treated as equal.
*   Strings are compared based on ordering in dictionary which is implemented via **ordinal values** (ASCII code or Unicode code). That is why `A` (ASCII code: 65) is less than `a` (ASCII cod: 97) as per their corresponding ASCII values.
*   Two same sequence objects are equal if they have same elements in same order.

| Sym | Description |
| --- | --- |
| == | Equals to: Returns True if the left operand and right operand has same values, otherwise False. |
| != | Not Equals to: Returns True if the left operand and right operand has different values, otherwise False. |
| > | Greater Than: Returns True if the left operand value is greater than right operand value, otherwise False. |
| < | Less Than: Returns True if the left operand value is less than right operand value, otherwise False. |
| >= | Greater Than or Equals to: Returns Trsue if the left operand value is greater than or equal to right operand value, otherwise False. |
| <= | Less Than or Equals to: Returns True if the left operand value is less than or equal to right operand value, otherwise False. |

````python
a=10
b=4
print("a==b : ",a==b)   # Equal 
print("a!=b : ",a!=b)   # Not Equal
print("a>b : ",a>b)     # Greater Than
print("a<b : ",a<b)     # Less Than
print("a>=b : ",a>=b)   # Greater Than or Equal To
print("a<=b : ",a<=b)   # Less Than or Equal To
````

### Identity operator
The operators `is` and `is not` are identity operators. 

* The `is` operator evaluates to `True` if the values of operands on either side of the operator point to the same object and `False` otherwise. 
* The `is not` operator evaluates to `False` if the values of operands on either side of the operator point to the same object and `True` otherwise.

````python
currentcity = "Una"
homecity = "Unao"
print(id(currentcity))
print(id(homecity))
print(currentcity is homecity)
print(currentcity is not homecity)
````

### Membership operator

Membership operator checks whether a value is present in a sequence (a string, a list, or a tuple, for membership) or not.  There are two membership operators in Python.

| Sym | Description |
| --- | --- |
| `in` | Returns `True` if the specified variable/literal is found, otherwise `False`. |
| `not in` | Returns `True` if the specified variable/literal is not found, otherwise `False`. |

````python
text = "Python is an easy and excellent language"
print("easy" in text)         # True
print("python" not in text)     # True

primes = [2,3,5,7,11,13,17,19]
print(12 in primes)
````

### Logical Operators

Logical operators are used to group two or more conditions. Python provides three logical operators to combine existing relational expressions.

| Sym | Description |
| --- | --- |
| `or` | Returns `True` any of the participating expressions results as `True`. |
| `and` | Returns `True` all the participating expressions results as `True`. |
| `not` | Returns `True` for `False` and `False` for `True`. |

````python
a = 10
b = 22
c = 4

c1 = a>b        # False
c2 = a>c        # True

print("c1 and c2:",c1 and c2)
print("c1 or c2:",c1 or c2)
print("not c1:", not c1)
````

### Bitwise Operators

Bitwise operators are used to perform bitwise calculations on integers. 

> The integers are first converted into binary and then operations are performed on each bit or corresponding pair of bits, hence the name bitwise operators. The result is then returned in decimal format.

| Sym | Bitwise | Description | Syntax |
| --- | --- | --- | --- |
| & | AND | Result bit 1,if both operand bits are 1, otherwise results bit 0. | x & y |
| \| | OR | Result bit 1,if any of the operand bit is 1, otherwise results bit 0. | x \| y |
| ~ | NOT | inverts individual bits | ~x |
| ^ | XOR | Results bit 1,if any of the operand bit is 1 but not both, otherwise results bit 0. | x ^ y |
| >> | Right Shift | The left operand’s value is moved toward right by the number of bits specified by the right operand. | x>>n |
| << | Left Shift | The left operand’s value is moved toward left by the number of bits specified by the right operand. | x<<n |

````python
a = 10  # 1010
b = 3   # 0011

print("a and b : ",a & b)  # and (&) performs a logical AND operation: 2
print("a or b : ",a | b)   # or (|) performs a logical OR operation: 11
print("a ^ b : ",a ^ b)    # xor (^) performs a logical XOR operation: 9
print("~a : ",~a)          # not (~) performs a logical NOT operation: -11
print("a << 2 : ",a << 2)  # Shift left by 2 bits: 40
print("a >> 2 : ",a >> 2)  # Shift right by 2 bits: 2
````

## Precedence and Associativity

### Operator precedence

> Operator precedence determines the way in which operators are parsed with respect to each other. Operators with higher precedence become the operands of operators with lower precedence. The operator precedence in Python is listed in the following table. It is in descending order (upper group has higher precedence than the lower ones).


| Operators | Meaning |
| --- | --- |
| () | Parentheses |
| ** | Exponent |
| +x, -x, ~x | Unary plus, Unary minus, Bitwise NOT |
| *, /, //, % | Multiplication, Division, Floor division, Modulus |
| +, - | Addition, Subtraction |
| <<, >> | Bitwise shift operators |
| & | Bitwise AND |
| ^ | Bitwise XOR |
| | | Bitwise OR |
| ==,!=, >, >=, <, <=, is, is not, in, not in | Comparisons, Identity, Membership operators |
| not | Logical NOT |
| and | Logical AND |
| or | Logical OR |

**Example** 
```python
print( 6 + 4 / 2)   # 8.0
print((6 + 4) / 2)  # 5.0
```

* Expressions with higher-precedence operators are evaluated first. In first expression multiplication / is having precedence over addition. So, 4 / 2 gets evaluated first and the result is added to 6.

* This behaviour has been overridden using parentheses as in second expression. Parentheses have the highest precedence and the expression inside the parentheses gets evaluated first, which in our case is (6 + 4) and the result is divided by 2.

````python
print( 6 + 4 / 2)   # 8.0
print((6 + 4) / 2)  # 5.0
````

### **Python operators’ Precedence Rule – PEMDAS**

Like the **BODMAS** rule in school’s mathematics class, Python uses a similar type of rule known as **PEMDAS**.
- Parentheses
- Exponentiation
- Multiplication
- Division
- Addition
- Subtraction

#### Associativity
> Associativity determines the way in which operators of the same precedence are parsed. Almost all the operators have left-to-right associativity.
```python
print(8 * 4 / 2)    # 16.0  
print(8 / 4 * 2)    # 4.0
```
In above example, for first expression, * and / has same precedence hence 4 and 8 are multiplied and afterwards product is divided by 2. In second expression first 8 is divided by 4 and then multiplied by 2. Hence, we have different results.

````python
print(8 * 4 / 2)    # 16.0  
print(8 / 4 * 2)    # 4.0
````
