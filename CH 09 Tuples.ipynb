## **TUPLE IN PYTHON**
---------
Tuple is another sequence data type like a list that allows store collection of any type values. But, unlike list, tuples are immutable (read-only) i.e, the elements of a tuple can't be changed in place .

### **FEATURES OF TUPLE**
-	Tuple can be homogeneous - hold same type of values
-	Tuple can be heterogenous to hold different types of values.
-	Tuples are immutable, i.e., individual element cannot be changed
-	Tuple element are accessible by their index 
    -	Index of first element is 0 (zero) (positive)
    -	Index of last element is -1 (negative)
-	Tuples are faster than list


### **Creating a Tuple - Tuple Display Construct**
---
A tuple can be created by enclosing values within a round bracket `()`, each value separated by a comma. The round bracket indicates the start and end of the tuple. This method for creating tuple is known as tuple display construct.

**Syntax:** `tpl = (val1, val2, …)`

```python
primes = (2,3,5,7,11,14)        # all integer values
wkdays = ('Mon','Tue','Wed','Thu','Fri','Sat','Sun')    # all string values
candidate = ('Ram',28,4580.00,False)    # all mixed values

print(primes,wkdays,candidate,sep='\n')
```

### **THE tuple () FUNCTION**

The built-in `tuple ()` constructor generates a tuple from specified sequence. The sequence can be a string, tuple or list itself. If the sequence is not specified, then an empty tuple is created.

**Syntax:** `var = tuple(sequence)`


```python
tptext = tuple('Python')        # from string
tplist = tuple([1,2,3,4])       # from list
tpempty = tuple()               # empty tuple

print(tptext)
print(tplist)
print(tpempty)
```

### **Creating Single element tuple**

Single element tuple is automatically get converted into scalar variable of type depending upon the assigned value. For example, in following code, that statement `t = (1)` hasn’t created a tuple, rather than an integer variable having 1 as its value.  

```python
sng_ele_list = [1]
print(sng_ele_list, "is of type",type(sng_ele_list))
```

```python
sng_ele_tpl = (1)
print(sng_ele_tpl,'is of type',type(sng_ele_tpl))
```

> To create a single element tuple, just put a comma after the element. 

```python
sng_ele_tpl = (1,)
print(sng_ele_tpl,'is of type',type(sng_ele_tpl))
```

## **ACCESSING ITEMS IN A TUPLE**
------

### **Using index**
Items in a tuple can be accessed by its index. Index of the first element of tuple is zero (0). Like list, items in tuple, can be accessed by positive or negative index.
 
-	`tpl [n]` will address element at (n+1)<sup>th</sup> position from left. For example, `prime[3]` represents 4<sup>th</sup> element from left that is 7.

-	`tpl [-n]` will address element at n<sup>th</sup> position from right. For example,` prime[-3]` represents 3<sup>rd</sup> element from right that is 7.

#### **Accessing Errors**

- `IndexError` is raised for invalid index specified while accessing tuple element.
- `TypeError` is raised in case floating point index is specified to accessing tuple element.


```python
print(primes)
print(primes[3])        # fourth item from left
print(primes[-5])       # fifth item from right  
```

### **TUPLE SLICING**

Like list, slicing extracts elements between start and end-1 position.

|  **style**  | **Description**                       |
|----|------|
|`tpl[0:n]` <br> `tpl[:n]`	|extract first `n` elements. For example: `prime [0:3]` will return items between index *0* and *2*|
|`tpl[n:]`|	extract all elements from `n` index. For example: `prime [3:]` will return items from *3rd* index till last element|
|`tpl[s:e]`|	extract all elements between `s` and `e-1` index.  For example: `prime [1:5]` will return items between index *1* and *4*|
 



```python
fruits = ('apple', 'pineapple', 'mango', 'orange', 'pear')
print(fruits[:])	    # same as print(fruits) – all items
print(fruits[:3])	    # first three items fruits[0:3]
print(fruits[3:])	    # all items after index 3 # fruits[3:5]
print(fruits[1:3])	    # two items – 2nd and 3rd 
print(fruits[-3:-1])	# negative slicing
```

### **TRANSVERSING TUPLE**

Transversing means accessing and processing individual element in a tuple in sequential order. This can be done using for loop.


```python
for fruit in fruits:
    print(fruit)
```

### **PACKING AND UNPACKING TUPLE**

- Tuple can be created by supplying values each separated by a comma, without round parentheses. This is called **packing**.

- When tuple element values are extracted into standard variables, it called **unpacking**. *Note that number of variables must be equal to the total no of elements in a tuple*.

```python
# packing example
pack_tuple = 2,4,6,8,10
print(pack_tuple, 'is of', type(pack_tuple))
```

```python
# unpacking example
t1 = (1,2,3)
a,b,c = t1
print(a,b,c)
```

```python
# unpacking example
t1 = (11,12,13)
a,b = t1[1:]
print(a,b)
```

### **TUPLE METHODS**

- `tpl.count(value)`: Returns frequency of an item in a Tuple
- `tpl.index(value)`: Returns position of the given value in the tuple, Otherwise, it will show `ValueError`.

```python
data = (9,2,1,2,4,3,1,5,3,2)
print(f"The value 2 repeats {data.count(2)} times")
print(f"The value 2 first appears at {data.index(2)}")
```

### **TUPLE FUNCTIONS**
| Function | Description |
|-----|----------------|
|`len (tpl)`| Returns total number of elements in the given Tuple|
|`max (tpl)`| Returns largest value in the tuple. It may raise `TypeError` exception  for incompatible datatypes. |
|`min (tpl)`| Returns smallest value in the tuple. It may raise `TypeError` exception  for incompatible datatypes.|
|`sum (tpl)`| Returns aggregate of tuple containing numeric values only, otherwise, it will raise `TypeError` exception.|
|`sorted(tpl)`| Arrange items in a tuple in ascending order (low to high) |
|`sorted(tpl, reverse=True)` | Arrange items in descending order (high to low)|

```python
t1 = (2.34,8.9, 2.67, 0.6,5.6)
print("Number of Items:",len(t1))
print("Maximum value:",max(t1))
print("Minimum value:",min(t1))
print("Sum of values:",sum(t1))
```

```python
print("Original Tuple:",t1)
print("Sorted in Ascending order:",sorted(t1))
print("Sorted in Descending order:",sorted(t1,reverse=True))
print("Original Tuple:",t1)
```
