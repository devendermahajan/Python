## **STATEMENTS IN PYTHON**
----------
Statements form the smallest executable unit within a Python program. Statements are the instructions given to the computer to perform any kind of action, be it data movements, and be it making decisions or be it repeating actions. Python statements can be classified as:

### **EMPTY STATEMENT**
-----------
The empty statement is the one which does nothing and moves to next statement in the flow of control. In Python an empty statement is pass statement. It can also be termed as null operation statement.

```python
pass
```
A pass statement is useful in those instances where the syntax of the language requires the presence of a statement but where the logic of the program does not.

``` py
def function_to_implemented():
    pass
```

### **SIMPLE STATEMENT**
-------
Any single line executable statement is a simple statement in Python. For example, following is a simple statement in Python:

```py
name = input ("Your name: ")
```
### **COMPOUND STATEMENT**
-------
A compound statement represents a group of statements executed as a unit. The compound statements of Python are written in a specific pattern as shown below :
```py
header-line:
    indented body of statements
```
- a header line which begins with a keyword and ends with a colon.
- a body consisting of one or more Python statements, each indented inside the header line. All statements in the body are at the same level of indentation.
- First unindented statement indicates the end of prev block.
- `if...else`, `match...case`, `while`, `for`,`def`, `with` are the statements that have multiple statements beneath them.

```py
def double(obj):        # header line
    return obj*2        # indented body of statments

print(double(6))
```

## **THE CONTROL FLOW**
--------
A program's control flow is the order in which the program's code executes. The control flow of a Python program is regulated by conditional statements, loops, and function calls.

### **SEQUENTIAL CONSTRUCT**
-----
Sequential flow refers to a linear approach wherein previous statement's output is used by next statement as input. Here, program code follows the line-by-line execution, in which the statements are executed sequentially, in the same order in which they appear in the program.

**Example:**
```py
a = 10
b = 20
c = a + b
print(c)
```

### **SELECTION CONSTRUCT**
---------
It is also called **decision construct**. A block of code is added to or removed from the sequence of execution conditionally. If a condition evaluates to `True`, a course-of-action is followed otherwise another course-of-action if followed.   

**Example:**
``` py
num = int(input("Enter any number: "))
if num % 2 == 0:
    print("Even value")
else:
    print("Odd value")
```

````python
num = int(input("Enter any number: "))  #22 #31
if num%2==0:        #22%2=0 T    # 31%2=1 F
    print("Even value")
else:
    print("Odd value")
````

### **ITERATION CONSTRUCT**
------
This control structure allows to execute a block of code repetitively either for finite no of times or till the exit condition is not met. 
The set-of-statements that are repeated and again is called the body of the loop. The condition on which the execution or exit of the loop depends is called the *exit-condition* or *test-condition*.

Example:
``` py
for x in range(1,11):
    print(x,'-',x*x)
```

````python
print('1 - 1')
print('2 - 4')
print('3 - 9')
print('4 - 16')
print("...")
print('10 - 100')
````

````python
for x in range(1,101):
    print(x,'-',x*x)
````

## **RELATIONAL EXPRESSION**
--------
This type of expression is used to compare two values and return either `True` or `False`. It also referred to as condition or boolean expression.

A relational expression can be create using following syntax:
``` py
expression1 <operator> expression2	
```
where `expression1` and `expression2` can be either variables, values, or some arithmetic expressions.

* `a==b` comparing two variables
* `a>2`	comparing the value of variable a with a literal.
* `a%2 == 0` arithmetic expression with a value. `0==a%2` have the same effect.

#### **RELATIONAL OPERATORS**
------
To create relational expression relational operators are used. Boolean expression is evaluated from Left to right i.e., left expression is compared with right expression.
- `==` equals to
- `!=` not equals to
- `>` Greater than
- `<` Less than
- `>=` Greater than or equals to
- `<=` Less than or equals to

````python
# Relational operator examples
a = 10
b = 15
print("a == b: ", a == b)   # False
print("a != b: ", a != b)   # True
print("a > b: ", a > b)     # False
print("a < b: ", a < b)     # True
print("a >= b: ", a >= b)   # False
print("a <= b: ", a <= b)   # True
````

> **Syntax error:** On using assignment operation (`=`) instead of equality operator i.e., instead of `a==b`, `a=b` is used. Interpreter will generate syntax error.

``` py

```
