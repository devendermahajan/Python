#  **THE if STATEMENTS.**
------------
The **`if`** statements are the conditional statements in Python and these implement selection constructs (decision constructs). Python supports three types of **`if`** statements:
- Simple `if`
- `if...else`
- Nested `if`
- `if...elif...else`
- shorthand `if`


## **THE IF STATEMENT**
------

- The `if` statement decides whether to run some statement or not depending upon the value of the boolean expression. If the boolean expression evaluates to `True` then statements in the if block will be executed; otherwise, none of the statements are executed. 
- The `if` block statements are determined through indentation.
- The `if` decision control flow statement starts with `if` keyword, followed by a boolean expression and ends with a colon (`:`). 
    ```py
    if expression:
        statement
        [statements]
    ```
The first unindented statement marks the end of the block. 

The `if` condition should contain multiple statements with the same indentation, otherwise it will raise an *`IndentationError`*.

````python
age = 8
# voting only on or after 18 year
if age>=18: # 18>=18 True   # 8>=18 False
    print("Yes you can vote.")
    print("First you have to register yourself for voter id")
    print("Be the part of biggest democratic festival.")
    print("You will be a nation builder")
print("Welcome")
````

## **THE IF…ELSE STATEMENT**
-----

The `if…else` statement allows for a two-way decision. An `if` statement can also be followed by an else statement. An else statement does not have any condition. Statements in the `if` block are executed in case the boolean expression is evaluated to `True` and `else` block statements are executed otherwised.

The syntax for `if…else` statement is,
```py
if expression:
    #if-block-statements 
else:
    #else-block-statements
```

````python
age = 8
# voting only on or after 18 year
if age>=18: # 18>=18 True   # 8>=18 False
    print("Yes you can vote.")
    print("First you have to register yourself for voter id.")
else:   # 8>=18 
    print("Wait till you are 18.")
    print("Learn the process of election.")
````

````python
# read two values and print the smallest of them
num1 = 32
num2 = 32

if num1<num2:   # 30<32 T  # 34<32 F  # 32<32 F
    print(num1,'is smallest')
else:
    print(num2, "is smallest")
````

## **NESTED IF STATEMENT**
When a program has an `if...else` statement inside another `if...else` statement it is called nesting. It can be done upto any level, indentation is the only way to figure out the level of nesting. 
The syntax for nested if…else statement is,
``` py
if conditon1:
    if condition12:
        statement1 
    else:
        statement2
else:
    else-block-statements
```
If the *condition1* is evaluated to `True`, then the control shifts to *condition2* and if it is evaluated to `True`, then *statement1* is executed, if it is evaluated to `False` then the *statement2* is executed. If the *condition1* is evaluated to `False`, then *else-block-statements* is executed.

````python
# read two values and print the smallest of them. If both are same, print "Equal"
num1 = 34
num2 = 32

if num1 == num2:     # 32==32 T    # 30==32 F       # 34==32 F
    print("Both are same.")
else:       # both are different
    if num1<num2:   # 30<32 T  # 34<32 F 
        print(num1,'is smallest.')
    else:
        print(num2, "is smallest.")
````

## **THE IF…ELIF…ELSE STATEMENT**
----
The `if…elif…else` is also called as multi-way decision control statement. It is best when logic must choose from several possible alternatives, then an `elif` statement is used along with an `if` statement. The keyword `elif` is short for else if. The `else` statement must always come last and will again act as the default action. 

The syntax for `if…elif…else` statement is,
``` py
if condition1:
    # if-block-statements-1
elif condition2:
    #elif-block-statement-2 
…
else:
    #else-block-statements
```
In the case of multiple conditions, only the first logical condition which evaluates to `True` will be executed. More specifically,
* If `condition1` is `True`, then if-block-statements are executed.
* If `condition1` is `False` and `condition2` is `True`, then `elif`-block-statement-1 is executed.
* If `condition1` and `condition2` are `False` and condition3 is `True`, then `elif`-block-statement-2 is executed and so on.
* If none of the conditions is `True`, then else-block-statements are executed.

````python
# test number is zero, positive or negative
n=int(input("Enter a number: "))

if n==0:
    print("zero")
elif n>0:
    print("positive")
else:
    print("negative")
````

### **THE SHORTHAND IF...ELSE**
------------
As single line of code is known as shorthand statement, Python supports many sorts of shorthand statements that help writing more compact code using less space.
The shorthand `if...else` is one of them. 

The syntax for shorthand `if…else` statement is,
``` py
true-statement if expression else false-statement
```
The condition is checked, a `True` condition takes the control of the program to the `true-statement` whereas a `False` condition takes it to the `false-statement` of the `if-else` shorthand.


````python
n=11
if n%2==0:  # 10%2==0 0==0 T  # 11%2==0 1==0 F
    print("Even")
else:
    print("odd")
````

````python
# Check whether a number is even or odd using shorthand if else
n = 4
print ("Even" if n % 2 == 0 else "Odd")
````

### **THE MATCH CASE STATEMENT**
----
Python `match` statements were introduced in *Python 3.10* and it is providing a good readability, and cleanliness in comparsion to the `if elif else` ladder statements. It like `switch...case` from languages like C/C++ or Java. 

**Syntax:**
```py
match expression:
    case matching_value_1: 
	statement_1
    case matching_value_2: 
	statement_2
…
    case matching_value_n: 
	statement_n
    case _: 
	nothing_match_case_statement
```

The `match` statement is initialized with the `match` keyword creating a block and taking an expression and then steps down to the various cases using the `case` keyword and the pattern, for the pattern to match the parameter. The *`_`* is the wildcard character which is run when nothing is matched.

````python
# # Check whether a number is even or odd using match expression
number = 61
match number % 2:   # 6%2=0  # 61%2=1
    case 0:
        print("Even")
    case 1:
        print("Odd")
````

> Use a pipe operator (`|`) to specify multiple values that a `case` can match

````python
# check whether a character is a vowel, consonent, digit, space 
# and another character using match statement
char = input("Enter a character: ").lower()[0]      # fetch only first character from input
match char:
    case 'a'|'e'|'i'|'o'|'u':
        print("Vowel")
    case ' ' | '\t' | '\n' | '\r':
        print("Whitespace")
    case char if 'a' <= char <= 'z':    #  char>='a' and char<='z'
        print('Consonent') 
    case char if '0' <= char <= '9':
        print("digit")
    case _:
        print("Other character")
````
