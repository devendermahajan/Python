## **INTRODUCTION**

A *set* object is an **unordered collection of distinct objects**. It is commonly used in membership testing, removing duplicates from a sequence, and computing mathematical operations such as intersection, union, difference, and symmetric difference. 

### **Features**
- Sets are unordered mutuable collection of unique elements. 
- Duplicate elements are automatically discarded.
- Heterogeneous objects are allowed.
- Indexing/slicing is not applicable.
- A set itself may be modified, but the elements contained in the set must be of an immutable type.


## **CREATING A SET**
------

### **1. Set Display Construct METHOD**
To create a list place the items inside a curly bracket `{}` each separated by a comma. The curly bracket marks the start and end of the set. This method for creating list is known as set display construct.

**Syntax:**  `var = {val1, val2, …}`

````python
# ordered collection such as list, tuple, string dictionary
l1 =[6, 1, 2, 5, 4, 3]
print(l1)
````

````python
# unordered: original position of value is not retained
setx = {6, 1, 2, 5, 4, 3}
print(f"{setx} is of {type(setx)}")
````

> In 'set' duplicate element are automatically discarded.

````python
setx = {10, 20, 10, 50, 30, 40}
print(setx)
````

### 2. The set() Function

The built-in `set ()` function can also be used to create a set  from specified sequence such as a string, tuple or list. If no sequence is specified, an empty set is created.

**Syntax:** `var = set([sequence])`

**Example**

````python
# set from a list
import random as ra 
l1 = [ra.randrange(5,30) for x in range (30)]       # collection 30 random values
print(l1)
````

````python
# a set from list
s1 = set(l1)
print(s1)
````

### **Element Immutability**
> A set  can contain any element of immutable type such as integer, float, tuple etc. But mutable elements (list, dictionary, set) can't be a member of set, and will raise an `TypeError` exception.

````python
# No errors
set1 = {1, "Ram", True, (3, 4), 5}      # used tuple as member (immutable)
print(set1)
````

````python
# TypeError: cannot use 'list' as a set element (unhashable type: 'list')
set2 = {1, "Ram", [3, 4], 5}            # used list as a member (mutable)
print(set1)
````

> - Items can added or removed from the set. This makes it mutable
> - But items can't be modified. This makes it immutable to editing

````python
s2 = {1,2,3}
print(s2)
````

````python
s2.add(4)
print(s2)
print(s2.pop(),s2,sep="\n")
````

````python
# TypeError: 'set' object does not support item assignment
s2[1]=30
````

### 3. **CREATING EMPTY SET**

Creating an empty set is a bit different because empty curly `{}` will result in an *empty dictionary*. Instead use the `set()` method without an argument to create an empty set.

````python
# empty sequences
el = []     # empty list
text =''    # empty string
et = ()     # empty tuple
print(type(el))
print(type(text))
print(type(et))
````

````python
setx = {}		# empty dictionary
print(f"{setx} is of {type(setx)}")
````

````python
# use set() to create empty set
setx = set()
print(f"{setx} is of {type(setx)}")
````

### **Fetching a set**
Since, sets do not support indexing, slicing, or other sequence-like behaviour.  Fetching element using index will raise TypeError exception.

````python
num_set = [0, 1, 2, 3, 4, 5]
print(num_set)
print(num_set[4])
````

````python
num_set = {0, 1, 2, 3, 4, 5}
print(num_set)
print(num_set[0])   # TypeError: 'set' object is not subscriptable
````

### **TRANSVERSING SET**
The items in a set can be transversed using a loop. However, since sets are unordered, it is undefined which order the iteration will follow.

````python
for n in num_set:
  	print(n)
````

## **SET MANIPULATIONS**

### **1. ADD MEMBER(S) IN PYTHON SET**
Python provides the `add()` method and `update()` method which can be used to add some particular item to the set. 


- ADD() METHOD
    The `.add()` method is used to single item to an existing set 
Example
```python
>>> color_set = set()		# A new empty set.
>>> color_set.add("Red")	# append a single member.
>>> print(color_set)
{'Red'}
```


````python
color_set = set()
color_set.add("Red")
print(color_set)
color_set.add('Green')
print(color_set)
````

UPDATE() METHOD
The .update() method is used to add multiple elements to the set.
Example
```python
>>> 
>>> color_set.update(["Blue", "Green"])	    # Add multiple items
>>> print(color_set)
{'Red', 'Blue', 'Green'}
```

````python
clr_lst = ['Blue','Orange','Yellow','Purple']       # list
color_set.update(clr_lst)
print(color_set)
````

### **2. REMOVE ITEM(S) FROM PYTHON SET**
The `pop()`, `remove()` and `discard()` functions are used to remove individual item from a Python set
- The `.pop()` method to remove the item from beginning. The `.pop()` method throws `KeyError: 'pop from an empty set'` when called upon an empty set
- The `.remove(value)` method removes the specified value from the set. If the specified value is not found, raise `KeyError` error
- The `.discard()` method removes a specific element from the set. If the item does not exist in the set then the set remain unchanged

````python
print(s1)
````

````python
# .pop()
print("Value removed:",s1.pop())
print("Remaining set:",s1)
````

````python
es = set()
es.pop()
````

````python
# .remove()
s1.remove(14)
print(s1)
````

````python
# .discard()
print(s1)
````

````python
s1.discard(20)
print(s1)
````

## **SET OPERATIONS**
- union
- intersection
- difference

````python
import random as ra 

a = {ra.randrange(5,25) for x in range(20)}
b = {ra.randrange(5,25) for x in range(20)}
print(a,b,sep='\n')
````

#### **1. Union**
> All elements from set1 and set 2 and single copy of common elements



```python
    set1.union(set2)
    set1 | set2
```

````python
# union
# set1 | set2 or set1.union(set2)
print("Using .union method: ", a.union(b))
print("Using bitwise or operator: ",(a|b))
````

#### **Intersection**
> All common elements from set1 and set 2



```python
    set1.intersection(set2)
    set1 & set2
```

````python
# intersection
print("Using bitwise & operator:", (a&b))
print("Using .intersection() methd:", a.intersection(b))
````

#### **Difference**
> All elements from the set1 which are not in the set2



```python
    set1.difference(set2)
    set1 - set2
```

````python
# difference a - b
print("Using - operator:", (a-b))
print("Using .difference() methd:", a.difference(b))
````

````python
# difference b - a
print("Using - operator:", (b-a))
print("Using .difference() methd:", b.difference(a))
````

#### **Symmetric Difference**
> All elements from the set1 and set2 leaving out common elements



```python
    set1.symmetric_difference(set2)
    set1 ^ set2
```

````python
print("Set 1:",a)
print("Set 2:",b)
````

````python
# Symmetric difference - Elements in a and b leaving common elements out
print("Using ^ operator:", (a^b))
print("Using .symmetric_difference() methd:", a.symmetric_difference(b))
````

#### **DISJOINT SETS**
Two set are called disjoint if both don’t have any element in common.  In Python, `A.isdisjoint(B)` returns `True` if A and B have no element in common

````python
# .isdisjoint returns True if sets are different otherwise false
s1 = {1,2,4}
s2 = {3,5,6}
print(s1.isdisjoint(s2))
````

````python
s1 = {1,2,4}
s2 = {4,1,2}
s1.isdisjoint(s2)
````

### **ISSUBSET**
A set A is considered a subset of another set B if every element of A is in B. In Python, `A.issubset(B)` and `A <= B` returns True if A is a subset of B.

```python
    set1.issubset(set2)
    set1 <= set2
```

````python
# a set A is a subset of another set B if all elements of set A are present in set B
a = {1,4,3}
b = {2,1,3}
````

````python
# subset
print("Using <= operator:", (a<=b))
print("Using .issubset() method:", a.issubset(b))
````

### **ISSUPERSET**
A set A is considered a superset of another set B if every element of B is in A. Returns True if A is a superset of B.

```python
    set1.issuperset(set2)
    set1 >= set2
```

````python
# superset
# a set A is a superset of another set B if all elements of set B are present in set A
print("Using >= operator:", (b>=a))
print("Using .issuperset() method:", b.issuperset(a))
````

### **Proper Superset**
> A set A is a proper superset of another set B if set A contains all elements of set B and have additional elements of its own

```python
    set1 > set2
```

````python
a = {1,3,4,2}
b = {3,1,4}
print("A is proper superset of B using > operator:", (a>b))
print("A is superset of B using >= operator:", (a>=b))
````
