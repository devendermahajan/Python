## Data Type
----
Data Type represent the type of data present inside a variable. Python is not static-type language. Hence, in Python, value stored in variable/constant, decides its Datatype.





  ### Number 
  *****
  Python has three types of numbers:
*   *Integer*: It does not have any fractional part. For example, `5` or `-5`.
    *   *Binary:* It can store an integer literal beginning with `0b` (digit zero followed by letter b). A binary value consisting of sequence of 0's and 1's. For example: `0b01001`.
    *   *Octal:* It can store an integer literal beginning with `0o` (digit zero followed by letter o). An octal value can contain only digit 0-7; 8 and 9 are invalid digits. For example:` 0o56`, `0o25`.
    *   *Hexadecimal:* It can store an integer literal beginning with `0x` or `0X` (digit zero followed by letter x). A hexadecimal value can contain only digit 0-9 and A (10)-F (15). For example: `0x56`, `0X2F7`.
*   *Floating Point*: It can store number with a fractional part. For example, `5.6` or `-5.6`. Floating point values can be written in two forms.
    *   *Fractional Form:* `35.76`, `0.0004` etc.
    *   *Exponent Form:* `3.50E03`, `0.5E-04` etc
*   *Complex*: It can store real and imaginary parts. Mathematically, a complex number is a number in the form `a + bj`, where `a` is real, and `b` represents imaginary component where `j` is equal to the square root of `-1`. For example: `4 + 5j`.

````python
# numbers
int_var = 1         # integer
float_var = 1.0     # float
complex_var = 1 + 2j # complex  (real + imaginary)
bin_var = 0b1010    # binary    
oct_var = 0o12      # octal
hex_var = 0x1a      # hexadecimal
print(int_var, 'is of type', type(int_var)) # 1 is of type <class 'int'>
print(float_var, 'is of type', type(float_var)) # 1.0 is of type <class 'float'>
print(complex_var, 'is of type', type(complex_var)) # (1+2j) is of type <class 'complex'>
print(bin_var, 'is of type', type(bin_var)) # 10 is of type <class 'int'>
print(oct_var, 'is of type', type(oct_var)) # 10 is of type <class 'int'>
print(hex_var, 'is of type', type(hex_var)) # 18 is of type <class 'int'>
````

## What is a variable?

- Variable can be defined as named location which stores values in main memory. 
- The value of variable can be used and processed during program execution (value in the variable may keep changing). For example, to store item name and its price, we need some identifiers. 
- In most programming languages, variables are used for storing User inputs, Intermediary Results, and Final Results.
- Variables in Python do not have fixed location. The location referred by variables changes every time their value change, though this is not true for all types of variables.

### Creating a Variable
Python has no such specific statement for declaring a variable. Variable is declared as soon as a value is assigned to it.

**Syntax:**

`var = value`

**Example:**
```python
lang = "Python" # string
age = 45 # integer
```

````python
lang = "Python"     # string variable
age = 45            # integer variable
print("I am learning", lang, "and I am", age, "years old") # I am learning Python and I am 30 years old
````

### Rules of identifier (variables, classes, functions, ...) naming

1. Variable name shall start with an **alphabet** or an **underscore**. Rest of variable name may contain alpha-numeric characters and underscores (`a-z`, `A-Z` `0-9`, and `_`)
    ```python
    var1 = 10 # valid variable name
    _var = 10 # valid variable name
    1var = 10 # variable name cannot start with number
    ```
2. Keywords cannot be using as variable name.
    ```python
    var1 = 10 # valid variable name
    pass = 10 # pass is a keyword cannot be used
    ```
3. Only permitted special character in variable name is underscore (`_`)
    ```python
    var 1 = 10 # invalid variable name space not allowed
    var_1 = 10 # valid variable name
    var$ = 10 # invalid special character not allowed
    ```
4. Variable name can be of any length.
5. Variable names are case sensitive i.e., `total` and `Total` shall be treated as different variables.
    ```python
    Total = 10 # valid variable name
    total = 10 # valid variable name
    ```

### Initialization
Assigning some value to the variable using assignment operator (`=`). A variable can be assigned with a fix value, another variable, or an expression.

1. **Literal**
    - `variable-name = value`
    -  Example: `a = 10`
2. **Another variable**:  
    - `variable-name = existing_variable`
    - Example: `b = a`
    > This does not create a new copy, rather both variables share common memory location where the value is stored.
3. **Expression**
    - `variable-name = expression`
    - Example: `c = a + b`
4. **Multiple assignment**:  We can assign different values to different variables in a single statement using following syntax:
    - `var1, var2, … = value1, value2, …`
    - Example: 
        ```python
        a, b, c = 1, 2, 3
        name, age, isalive = "Rahul", 25, True
        ```
5. Moreover, we can assign same value to multiple variables in single statement, as shown below:

   `a=b=c=d=e=0`

* **rvalues** are the literals, variables, or expressions is assigned to the lvalues. These appear on the right-hand side of an assignment operator.

*  **Strong typing** means that variables do have a type and that the type matters when performing operations on a variable.
*  **Dynamic typing** means that the type of the variable is determined only during runtime.

## Variable Internals
Every Python object has three key attributes associated to it:
### type
The type of an object determines the operations that can be performed on the object. The `type (object)` returns the type of an object.
```python
a = 10
type(a)
<class 'int'>
a = "Hello World!"
type(a)
<class 'str'>
```
### value
It is the data-item stored in the object. If it is literal itself or result of expression assigned to a variable. The `print ()` statement can be used to display the value stored in the variable.
```python
a = 10
print(a)
10
a = "Hello World!"
print(a)
Hello World
```
### id
The id of an object is generally the memory location of the object. The `id()` function returns the id of an object.
```python
print(id(10))
140735974204392
a = 10
print(id(a))
140735974204392
```
Here the object `10` is internally stored at location `140735974204392`. The variable `a` is current referencing location `140735974204392`. Hence it is clear that variable is not a storage location in Python, rather a label pointing to a value object.

## Mutable and immutable types
The Python data objects are classified as mutable and immutable types.

### Immutable types
Immutable types are those who do not allow to change value stored in the variable in place. Datatypes – _integers_, _floating point numbers_, _Booleans_, _strings_, _tuples_ are immutable.
In Python, variable names are references to data values. That means, variables names do not store any value themselves. So whenever values of an immutable variable is changed, it simply changes it reference to the address of new immutable value.
```python
a = 15
print(id(a))
20111651888400
a = "Hello World!"
print(id(a))
20111651887219
```
In above example, it may be noticed that as value of variable a changed to `"Hello World"` from `15`, its address also changes. Thus, we can justify immutability of types.

### Mutable types
The mutable types are those whose values can be changed in place. Python has only three types that are mutable: _lists_, _dictionaries_, and _sets_.
```python
a = [1,2,3]
print(id(a))
15111651889700
a [1] = 4
print(id(a))
15111651889700
```
In above example, we changed the value of the item at index `1` in list `a`, but even after that the address of the list does not change.

## Type Casting

> The conversion of one data type into the other data type is known as type casting in python or type conversion in python. Python supports a wide variety of functions or methods.

### Implicit type conversion (Coercion).

An implicit type conversion is a conversion performed by the compiler without programmer's intervention. An implicit conversion is generally applied whenever differing data types are intermixed in an expression (mixed mode expression), so as not to lose information. In a mixed arithmetic expression, Python converts all operands up to the type of the largest operand (type promotion).
```python
a = 10
b = 2.5
print(type(a))
<class 'int'>
print(type(b))
<class 'float'>
c = a * b
print(type(c))
<class 'float'>
```

````python
a = 10
b = 2.5
print(type(a))      # <class 'int'>
print(type(b))      # <class 'float'>  
c = a + b
print("Sum of",a,"and",b,"=",c)      # <class 'float'>
print(type(c))
````

### Explicit Type Conversion
An explicit type conversion is user-defined conversion that forces an expression to be of specific type. The explicit type conversion is also known as _Type Casting_

#### Complex type conversion
> The built-in function `complex()` to converts values from other types to the `complex` type. Value can be any type including of `int`, `float`, `bool`, `str`. The complex function has the following two forms for conversion.
* `complex(x)` - Converts a value `x` into a `complex` type. In this form, the real value is `x`, and the imaginary value is `0`.
    ```python
    num = 45
    type(num)
    <class 'int'>
    cn = complex(num)
    type(cn)
    <class 'complex'>   
    ```
* `complex (x, y)` - Convert the value `x` and `y` into a `complex` type. In this form, the real value is `x`, and the imaginary is `y`.
    ```python
    num, img = 45, 5
    cnum = complex (num, img)
    type(cnum)
    <class 'complex'>
    ```

````python
# complex(x) : x can be a integer, float, bool, str etc.
num = 10
# print(type(num))    # <class 'int'>
cnum = complex(num) # (10+0j)
print(cnum,'is of type', type(cnum))   # <class 'complex'>

# complex(x, y) : x and y can be integer, float, bool, str etc.
n,i=10,2
cnum = complex(n,i) # (10+2j)
print(cnum,'is of type', type(cnum))   # <class 'complex'>

# conversion of string to complex
s = "10+2j"
print(type(s))      # <class 'str'>
cnum = complex(s)   # (10+2j)
print(cnum,'is of type',type(cnum))   # <class 'complex'>
````
