# **LIST IN PYTHON**

> List is the most frequently used data structures THAT allow to store collections of data in a **mutable** and **ordered** sequence. Each element or value that is inside a list is called an **item**. All the items in a list are assigned to a single variable. Lists are the most flexible data storage formats in Python because they can have values added, removed, and changed

## **FEATURES OF LIST**
----------
-	List can be homogeneous – one that holds same type of values.
-	List can be heterogenous – one that holds values of mixed type.
-	List can be of variable length.
-	List can be nested – one that can have another list within itself.
-	Lists are mutable, as its individual elements can be change in place without have the memory address of list changed.
-	List elements are accessible by their index (position in the list).
	* Index of first element is 0 (zero) (positive)
	* Index of last element is -1 (negative)

### **3. LIST COMPREHENSIONS**
- `a = {2,4,6,8,10...,20}`
- `a = {x : 2<=x<=20, x is an even number}`

List comprehension is an elegant and concise approach of making lists. With this, one could rewrite the for loop in just a single line of code. List comprehension is considerably faster than processing a list using the for loop.
Syntax:  
```py
var = [expression for element in iterable [if condition]]
```
Every list comprehension in Python includes three elements:
-	*expression* is the member itself, a call to a method, or any other valid expression that returns a value. 
-	*element* is the object or value in the iterable. 
-	*iterable* can be a list, set, sequence, generator, or any other object that can return its elements one at a time. 

For example, suppose we want to populate a list of squares of first 10 natural numbers. We will be using standard method as given below

````python
square_list = [1,4,9,16,25,36,49,64,81,100]
print(square_list)	
````

## **ACCESSING THE ITEMS OF A LIST**

Each item in a list can be called individually, through indexing. Lists use square brackets `[]` to access individual items. The index provided within the square brackets indicates the value being accessed.

**Syntax:** `list_obj[index]`

- Python uses *positive indices* to access the list elements from the left side. An index of `0` would access the first element of the list, `1` second, `2` third, and so on.

- Likewise *negative indices* can be used to access list items from right side.  An index of `-1` would access the last element of the list, `-2` second last, `-3` third last and so on.


````python
list_a = [1, 2, 3, 4, 5]
print(list_a)
````

````python
# accessing first items using positive index
print("First Element:",list_a[0])        # first item

# accesing last item using positive index
last_index = len(list_a)
print("Last Element:",list_a[last_index-1]) # last item
````

````python
# accessing last item using negative index
print("Last Element:",list_a[-1])       # last item

# accessing first item using negative index
print("First Element:",list_a[-last_index])      
````

````python
# find the smallest and largest number in a list using indexing
numbers = [5, 2, 9, 1, 35, 6]
smallest = numbers[0]  # Assume the first number is the smallest
largest = numbers[0]   # Assume the first number is the largest

for n in numbers:
    if n < smallest:
        smallest = n
    if n > largest:
        largest = n

print("Smallest:", smallest)
print("Largest:", largest)
````

````python
numbers.sort()  # Sort the list in ascending order
print("Smallest:", numbers[0])  # First item is the smallest    
print("Largest:", numbers[-1])  # Last item is the largest
````

### **enumerate() Function**
The `enumerate()` function a counter to an iterable (list, string, tuples) and returns it as an *enumerate object*. This object can then be used in a loop to access both the *index* and the *value* of each item in the iterable simultaneously. 

**Syntax:**
`enumerate(iterable, start=0)`

- `iterable`: Any object that supports iteration (e.g., list, tuple, string, set).
- `start`: An optional integer argument that specifies the starting index for the counter. By default, it is 0.

For each item, it generates a pair as tuple: the current *index* (starting from `start`) and the *item* itself.

````python
my_list = ['apple', 'banana', 'cherry']

# Looping with enumerate()
print("With default start index:")
for index, value in enumerate(my_list):
    print(f"Item {index}: {value}")
````

````python
# Looping with a custom start index
print("With custom start index:")
for index, value in enumerate(my_list, start=1):
    print(f"Item {index}: {value}")
````

## **LIST SLICING**
----
> List slicing extracts a part of the list by specifying index range along with the colon (:) operator which itself is a list. 

Syntax: 
```py
seq_obj [[start=0]:[end=len][:step=1]]
```
- where both `start` and `end` are integer values (positive or negative values). 
- List slicing returns a part of the list from the `start` index value to `end` index value which includes the `start` index value but excludes the `end` index value. 
- The optional `step` specifies the increment value to slice by.


````python
fruits = ['apple', 'pineapple', 'mango', 'orange', 'pear','plum','grapes','banana']
for index,name in enumerate(fruits):
    print(f"({index}, {name})")
````

````python
print (fruits)
````

````python
# same as print(fruits) – all items
print(fruits[:]) # print(fruits)
````

````python
# first three items fruits[0:3]
print(fruits[:3])       # start=0 end=3
````

````python
# all items after index 3 # fruits[3:8]	    
print(fruits[3:])   # start=3 end=8
````

````python
# four items – 2nd to 4th element     
print(fruits[1:5])  # start=1 end=5 
````

````python
# negative slicing   
print(fruits[-3:-1])	
````

````python
print(fruits[-3:])  # last three elements
````

### **STEP IN SLICE OPERATION**
In the slice operation, third optional argument, `step` refers to the number of items that can be skipped after the start indexing. The default value of step is one (if not specified). 

````python
print(fruits[:])        # all items
print(fruits[::2])      # start=0, end=len, inc=2 => 0 2 4 6 8
print(fruits[1::3])     # start=1, end=len, inc=3 => 1 4 7 10
````

````python
# reverse a sequence
print(fruits[::-1]) # start=0, end=len, inc=-1
````

#### 2. **`.extend()` METHOD**

The `extend ()` method is used for adding multiple elements, given in form of a list to a list. It takes a list as an argument and appends all the elements of argument list to the list object.

**Syntax**: `list_object.extend(iterator)`


````python
new_clr = list(colours) # creating a new list new_clr with the same items as colours
new_colours = ['white', 'black']    # creating a new list new_colours with two items
print(new_clr)          # printing new_clr to show that it has the same items as colours
new_clr.append(new_colours) # appending new_colours as a single item to the end of new_clr
print(new_clr)  # printing new_clr to show that it now has new_colours as a single item at the end
````

````python
print(colours)              # printing colours to show that it is unchanged
new_colours = ['white', 'black']    # creating a new list new_colours with two items
colours.extend(new_colours)     # extending colours by adding items of new_colours to the end of colours
print(colours)      # printing colours to show that it now has items of new_colours added to the end
````

##### **Example**
In this example, let us find the month with the smallest and largest consumer price index (CPI).

To identify the month with the smallest consumer price index, 
- Apply the `min()` function on prices list to identify the min_price. 
- Apply the `max()` function on prices list to identify the max_price.

Next, use the `.index()` method over months list to find the index location of the min_price and max_price. 

Using this indexed location on months, you can identify the month with the smallest consumer price index.

````python
months = ['January', 'February', 'March', 'April', 'May', 'June']
prices = [208.11, 237.81, 438.91, 334.50, 123.54, 304.60]
````

````python
# Identify min and max price
min_price = min(prices)                 # 123.54
max_price = max(prices)                 # 438.91

# Identify min and max price index
min_index = prices.index(min_price)     # 4
max_index = prices.index(max_price)     # 2


# Identify the month with min price and max price
min_month = months[min_index]
max_month = months[max_index]

print(f"Minimum price {min_price} was in {min_month}")
print(f"Maximum price {max_price} was in {max_month}")
````

#### `filter()` function
For the `filter()` function, it takes a function and the lists and returns an iterator of filtered elements. You can either create a Python function or use a lambda function to get a filtered list. 

**Syntax:** `filter(func,list)`

- `func` parameter is a function that will return boolean values.  
- `list` parameter is the list on which filter shall be appiled.

**Example:**


In our case, we will filter item prices greater than 350. Then, we will use `filter()` to apply the function on item_price list and get an iterator of filtered elements. 

To access the filtered elements, we can either extract value using a for loop or convert an iterator into the lists.  


````python
item_price = [ra.randrange(250,751) for x in range(40)]
print(item_price)
````

````python
gt_price=[]
for pr in item_price:
    if pr>350:
        gt_price.append(pr)

print(gt_price)
````

````python
# list comprehension
fl_price = [x for x in item_price if x>350]
print(fl_price)
````

````python
# Using lambda function and get filtered elements.
# fun_obj = lambda args : expression
rule = lambda a: a > 350    # True or False
````

````python
filtered_price = list(filter(rule, item_price))

print(filtered_price)
````

### **SHALLOW COPY**

A bitwise copy of an object, where a new object is created and it has the same copy of the values in the original object, is called a shallow copy. It creates a new collection object and then populating it with references to the child objects found in the original. In essence, a shallow copy is only one level deep. 

A *shallow copy* means any modification in the new list won’t be reflected in the original list. 


#### **Using `list()` constructor**

````python
l1 = [1, 2, 3]
l2 = list(l1) 
print("ID of both list after using list():",id(l1), id(l2),sep="\n")
print("Contents of both list before modifications:",l1,l2,sep="\n")
````

````python
# adding new item to l1
l1.append(4)	
print("Contents of both list after modifications:",l1,l2,sep="\n")
````

#### **Using `.copy()` Method**

````python
l1 = [1, 2, 3]
l2 = l1.copy() 
print("ID of both list after using .copy() method:",id(l1), id(l2),sep="\n")
print("Contents of both list before modifications:",l1,l2,sep="\n")
````

````python
# adding new item to l1
l2.append(4)	
print("ID of both list after modifications:",id(l1), id(l2),sep="\n")
print("Contents of both list after modifications:",l1,l2,sep="\n")
````

### **DEEP COPY**

A deep copy creates a new object and recursively copies all objects found in the original, ensuring that changes to the new object do not affect the original object and vice versa. Copying an object this way walks the whole object tree to create a fully independent clone of the original object and all of its children.

#### **How to?**
1. First import the `copy` module.
2. Use the `deepcopy()` function defined in the `copy` module.


````python
import copy                     # activating copy module
l1 = [1, 2, 3]                  
l2 = copy.deepcopy(l1)          # making deep copy of the list
print("Ids of list objects before changes:")
print(id(l1), id(l2),sep="\n")
print("Data in the lists before changes:")
print(l1,l2,sep="\n")
````

````python
l1.append(4)
l2[1] = 20	
print("Data in the lists after changes:")
print(l1,l2,sep="\n")
````

## **ACCESS ELEMENTS OF NESTED LISTS**
----
Accessing elements in a nested lists requires two indices: one for the outer list (rows) and another for the inner list (columns).

````python
def display(matrix):
    print("[")
    for rows in matrix:
        print("   [",end="")
        for cols in rows:
            print (cols,"",end = '')
        print("]")
    print("]")
````

````python
matrix = [
    	[1, 2, 3],
    	[4, 5, 6],
    	[7, 8, 9]
	]

display(matrix)
````

````python
# Here the interpreter will retrieve 3rd value from the 2nd row (as the index starts from 0 so [1] means “2” and [2] means “3”).
# arr[r][c]
print(matrix[1][2])
print(matrix[-1][-1])
````

## **FLATTEN A NESTED LIST**
---

Flattening a nested lists in Python transforms it into a single list containing all of its elements. There are various methods to convert a nested list into a flat list. 

Let us explore them in following sections. 

### **1. USING FOR LOOP**

Although nested list can be flatten using for loop, but this method is not recommend. 

````python
flatlist = []
for row in matrix:
    for element in row:
        flatlist.append(element)

print(flatlist)
````

### **2. Using List Comphrension**
List comprehension is the simplest method  for flatting a nested list  with the second argument set to empty square brackets [].

````python
flatlist = [val for row in matrix for val in row]
print(flatlist)
````

### **3. USING SUM FUNCTION**

The `sum ()` can also help flatting a nested list. It requires two arguments to flatten the nested list: a nested list and an empty list (i.e. `[]`). The `sum()` combines nested list and `[]` to produce a flattened list.

````python
flatlist = sum(matrix,[]) 
print(flatlist)
````
